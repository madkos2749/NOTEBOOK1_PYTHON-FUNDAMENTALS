LOGGING:

WHAT IS LOGGING 

--> Logging is way to record what program is doing while it runs. 
--> Instead of using print() which we have to delete later, logging creates proper records with levels like DEBUG, INFO, WARNING, ERROR, CRITICAL and also saves to file.

```python
# Logging:
import logging

# Setup logging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    datefmt='%H:%M:%S'
)

def process_parcel(courier_name, parcel_weight):
    logging.info(f"{courier_name} started processing {parcel_weight}kg parcel")
    
    if parcel_weight > 20:
        logging.warning(f"Heavy parcel: {parcel_weight}kg by {courier_name}")
    if parcel_weight <= 0:
        logging.error(f"Invalid weight: {parcel_weight} by {courier_name}")
        return
    
    logging.info(f"Parcel delivered successfully by {courier_name}")

process_parcel("Harshit", 12)
process_parcel("Priyanka", 25)
process_parcel("Aruna", 0)
```

What program does: 
tracks courier parcel delivery with logs.

Which part code demonstrates:
logging.info(), warning(), error() demonstrates concept.

How it works: 
basicConfig sets level and format. Program runs and creates time-stamped logs with levels, better than print for debugging in production.

WHY USE LOGGING:

Why we need logging:

--> print() is temporary and disappears. 
--> Logging stays forever, shows exact time, shows severity level, and can be saved to file even after notebook closed. 
--> In real projects we cannot use 100 prints to find bug.
--> We use logging because:Track production issuesDon't stop program like print debuggingLevels help filter important vs normal messagesWorks even when user not watching screen.

```python
# Why Logging is Needed:
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format='%(levelname)s - %(message)s'
)

def check_atm_transaction(client_name, balance, withdraw):
    logging.info(f"Transaction started for {client_name}")
    logging.debug(f"Balance={balance}, Withdraw={withdraw}")
    
    if withdraw > balance:
        logging.error(f"Insufficient funds for {client_name}: need {withdraw}, has {balance}")
        return "Failed"
    
    if balance - withdraw < 500:
        logging.warning(f"Low balance alert for {client_name}: {balance-withdraw} left")
    
    logging.info(f"Transaction success for {client_name}")
    return "Success"

check_atm_transaction("Harish", 5000, 1000)
check_atm_transaction("Divya", 2000, 1800)
check_atm_transaction("Mohan", 1000, 2000)
```

What program does: 
simulates ATM withdrawal and shows why logging needed over print.

Which code part demonstrates:
INFO for normal flow, DEBUG for values, WARNING for low balance, ERROR for failure.

How it works:
Without logging we would not know why Divya got warning or Mohan failed. Logs give complete history with level, helps to fix bug later even without watching program live.

Difference Between print() vs logging():

print() vs logging:

PRINT(): print() is for user to see output now. 

LOGGING(): logging is for developer to record what happened with time and severity.print() always prints, cannot hide, cannot save to file properly, no level. logging you can turn off DEBUG in production, keep only ERROR, and save to file automatically.

```python
# print() vs logging:

import logging

logging.basicConfig(level=logging.WARNING, format='%(levelname)s: %(message)s')

def calculate_bus_fare(passenger_name, distance):
    # print() - always shows, good for user
    print(f"Hello {passenger_name}, calculating fare...")
    
    # logging - only shows if level matches, good for developer
    logging.debug(f"DEBUG: distance={distance} for {passenger_name}")
    
    if distance <= 0:
        logging.error(f"Invalid distance {distance} for {passenger_name}")
        print("Invalid distance!")
        return
    
    if distance > 100:
        logging.warning(f"Long trip {distance}km for {passenger_name}")
    
    fare = distance * 5
    print(f"Your fare is Rs.{fare}")
    logging.info(f"Fare calculated Rs.{fare} for {passenger_name}")

calculate_bus_fare("Suresh", 10)
calculate_bus_fare("Kavya", 150)
calculate_bus_fare("Ravi", -5)
```

What program does: 
calculates bus fare, shows both print and logging together.

Which code part demonstrates:
print() shows to passenger always. logging.warning/error only shows when problem, debug/info hidden because level is WARNING.

How it works:
print() cannot be filtered, but logging level WARNING hides DEBUG and INFO automatically. In production you set level to ERROR, all extra prints would still clutter, but logs stay clean and can be saved to file later.

LOG LEVELS:

--> Log levels tell how serious a message is. 
5 main levels in order:
Log levels tell how serious a message is.

5 main levels in order:
1. DEBUG = small detail, for developer only.
2. INFO = normal working.
3. WARNING = something unusual but program still runs.
4. ERROR = problem, some function failed.
5. CRITICAL = very big problem, program may stop.

```python
# Log Levels:

import logging

logging.basicConfig(
    level=logging.DEBUG,
    format='%(levelname)s - %(message)s'
)

def monitor_bakery_shop(baker_name, oven_temp):
    logging.debug(f"Checking oven sensor for {baker_name}: {oven_temp}C")
    logging.info(f"Bakery opened by {baker_name}")
    
    if oven_temp > 220:
        logging.warning(f"Oven hot {oven_temp}C for {baker_name}")
    if oven_temp > 300:
        logging.error(f"Oven overheat {oven_temp}C! Baking failed for {baker_name}")
    if oven_temp > 400:
        logging.critical(f"FIRE RISK {oven_temp}C! Shop closing for {baker_name}")
        return

monitor_bakery_shop("Anil", 180)
monitor_bakery_shop("Geeta", 250)
monitor_bakery_shop("Vijay", 350)
monitor_bakery_shop("Leela", 450)
```

LOG FORMATTERS:

--> Log Formatter controls how log message looks. 
--> By default log only shows message, but formatter can add time, file name, level name, function name etc. 
--> It makes log readable and useful for debugging.
--> Format uses placeholders like %(asctime)s, %(levelname)s, %(message)s.

```python
# Log Formatters:

import logging

# Create custom formatter
formatter = logging.Formatter(
    fmt='%(asctime)s | %(levelname)s | %(funcName)s | %(message)s',
    datefmt='%H:%M:%S'
)

# Setup handler with formatter
handler = logging.StreamHandler()
handler.setFormatter(formatter)

logger = logging.getLogger("HotelLogger")
logger.setLevel(logging.INFO)
logger.handlers.clear()
logger.addHandler(handler)

def book_room(manager_name, room_no):
    logger.info(f"{manager_name} booking room {room_no}")
    if room_no > 100:
        logger.warning(f"Room {room_no} on top floor for {manager_name}")

def cancel_room(manager_name, room_no):
    logger.error(f"{manager_name} cancelled room {room_no}")

book_room("Nisha", 45)
book_room("Tarun", 105)
cancel_room("Nisha", 45)
```

What program does:
books and cancels hotel rooms with formatted logs.

Which code part demonstrates:
logging.Formatter(fmt='...') is formatter concept, and handler.setFormatter(formatter) applies it.

How it works:
Formatter template adds time asctime, level, function name funcName, and message. So we know when, what level, from which function log came, not just message.

FILE HANDLERS:

--> File Handler sends log messages to a file instead of console. 
--> In Jupyter console logs disappear after closing, but file stays forever. 
--> We create FileHandler with file name, attach formatter, attach to logger. 
--> Then all logs go to that file for future audit.

```python
# File Handlers:

import logging

# Setup logger with file handler
logger = logging.getLogger("StoreLogger")
logger.setLevel(logging.INFO)
logger.handlers.clear()

file_handler = logging.FileHandler("store_logs.txt", mode='w')
formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s', datefmt='%H:%M:%S')
file_handler.setFormatter(formatter)
logger.addHandler(file_handler)

def sell_product(seller_name, product, quantity):
    logger.info(f"{seller_name} sold {quantity} x {product}")
    if quantity > 10:
        logger.warning(f"Bulk sale {quantity} by {seller_name}")
    if quantity <= 0:
        logger.error(f"Invalid quantity {quantity} by {seller_name}")

sell_product("Ishita", "Notebooks", 5)
sell_product("Karan", "Pens", 15)
sell_product("Ishita", "Eraser", 0)

# Read file to show in notebook
print("Logs saved to file:")
with open("store_logs.txt", "r") as f:
    print(f.read())
```

What program does:
sells store products and saves logs to store_logs.txt.

Which code part demonstrates:
FileHandler("store_logs.txt") is file handler concept.

How it works:
Instead of StreamHandler which prints to console, FileHandler writes to file. Formatter formats it. After run we open file and see all history saved, useful for checking later even after notebook closed.

CONSOLE HANDLERS:

--> Console Handler prints log messages to console or terminal or Jupyter output. 
--> It is opposite of File Handler. 
--> When we want to see logs live while program runs, we use StreamHandler which by default is console. 
--> We can control which level goes to console.

```python
# Console Handlers:

import logging
import sys

logger = logging.getLogger("MetroLogger")
logger.setLevel(logging.DEBUG)
logger.handlers.clear()

# Console handler - prints to screen
console_handler = logging.StreamHandler(sys.stdout)
console_handler.setLevel(logging.INFO)
formatter = logging.Formatter('%(levelname)s | %(message)s')
console_handler.setFormatter(formatter)
logger.addHandler(console_handler)

def check_metro_train(operator_name, signal_status):
    logger.debug(f"Debug check by {operator_name}: {signal_status}")
    logger.info(f"{operator_name} checking signal {signal_status}")
    
    if signal_status == "yellow":
        logger.warning(f"Slow down - Yellow signal seen by {operator_name}")
    if signal_status == "red":
        logger.error(f"STOP - Red signal! Train halted by {operator_name}")

check_metro_train("Sneha", "green")
check_metro_train("Aman", "yellow")
check_metro_train("Ritu", "red")
```

What program does:
monitors metro train signals and shows live logs on console.

Which code part demonstrates:
StreamHandler(sys.stdout) is console handler concept, setLevel(INFO) filters.

How it works:
Handler sends logs to Jupyter console. DEBUG log is hidden because handler level is INFO. Only INFO and above appear live on screen, no file created. Good for real-time monitoring.

ROTATING LOGS [Bonus]:

--> Rotating Logs means log file should not become too big. 
--> When file reaches some size, Python automatically creates new file and moves old logs to backup. 
--> We use RotatingFileHandler with maxBytes and backupCount. 
--> This prevents disk full in production.

```python
# Rotating Logs:

import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("CourierLogger")
logger.setLevel(logging.INFO)
logger.handlers.clear()

# Rotate after 300 bytes, keep 2 backup files
rotating_handler = RotatingFileHandler(
    "courier_logs.log", maxBytes=300, backupCount=2
)
formatter = logging.Formatter('%(asctime)s - %(message)s', datefmt='%H:%M:%S')
rotating_handler.setFormatter(formatter)
logger.addHandler(rotating_handler)

def dispatch_parcel(agent_name, parcel_id):
    logger.info(f"{agent_name} dispatched parcel {parcel_id}")

for i in range(10):
    dispatch_parcel("Prateek", f"PKG00{i}")

# Check created files
import os, glob
print("Created log files:")
for file in glob.glob("courier_logs.log*"):
    size = os.path.getsize(file)
    print(f"{file} - {size} bytes")
    with open(file) as f:
        print(f.read())
    print("---")
```

What program does:
dispatches 10 parcels and logs each, with rotation.

Which code part demonstrates:
RotatingFileHandler(maxBytes=300, backupCount=2) is rotating concept.

How it works: When file exceeds 300 bytes, it renames to .1, .2 and starts new file. Only 2 backups kept, so disk never full. Good for long-running services like courier tracking.

LOGGING BEST PRACTICES:

Logging best practices means; writing logs that help in real projects:
1. Never use print, use logger with name.
2. Use correct level, not everything as INFO.
3. Use formatter with time and function name.
4. Use try-except to log exceptions with exc_info=True.
5. Use file handler + console handler both.
   This makes debugging easy without changing code.

```python
# Logging Best Practices:

import logging

logger = logging.getLogger("PaymentLogger")
logger.setLevel(logging.DEBUG)
logger.handlers.clear()

# Console for user, File for record
console = logging.StreamHandler()
file_h = logging.FileHandler("payment.log", mode='w')

formatter = logging.Formatter('%(asctime)s | %(levelname)s | %(name)s | %(message)s')
console.setFormatter(formatter)
file_h.setFormatter(formatter)

console.setLevel(logging.WARNING)  # console shows only warning+
file_h.setLevel(logging.DEBUG)     # file saves everything

logger.addHandler(console)
logger.addHandler(file_h)

def process_payment(customer_name, amount):
    logger.info(f"Payment started for {customer_name}: Rs.{amount}")
    logger.debug(f"Validating card for {customer_name}")
    
    try:
        if amount <= 0:
            raise ValueError(f"Invalid amount {amount}")
        if amount > 50000:
            logger.warning(f"High value transaction Rs.{amount} for {customer_name}")
        
        logger.info(f"Payment success for {customer_name}")
        
    except Exception as e:
        logger.error(f"Payment failed for {customer_name}", exc_info=True)

process_payment("Divya", 2000)
process_payment("Harish", 75000)
process_payment("Meena", -100)

print("\n--- File content ---")
with open("payment.log") as f:
    print(f.read())
```

What program does:
processes payments with proper logging.

Which code part demonstrates:
Best practices: named logger PaymentLogger, two handlers with different levels, formatter with time/name, exc_info=True for exception, correct level usage.

How it works:
Console shows only important WARNING/ERROR to not clutter user, but file saves all DEBUG/INFO for developer audit later. Using exc_info=True saves full error traceback automatically.
