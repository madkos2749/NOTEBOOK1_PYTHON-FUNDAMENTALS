FILE HANDLING:

File Handling means; 
--> Reading text files means opening a .txt file and getting data from it into Python. 
--> We use open(file, "r") which is read mode. 
--> File pointer starts at beginning and reads line by line or full text. 
--> Always close file or use with open() which auto closes. 
--> If file not found, Python gives error.

```python
# Reading Text Files:

# First create a sample file (for demo)
with open("bus_passengers.txt", "w") as f:
    f.write("Raj - Seat 12\n")
    f.write("Simran - Seat 14\n")
    f.write("Arjun - Seat 15\n")

# Now reading concept
print("Full read:")
with open("bus_passengers.txt", "r") as reader:
    content = reader.read()
    print(content)

print("Line by line read:")
with open("bus_passengers.txt", "r") as reader:
    for line in reader:
        name, seat = line.strip().split(" - ")
        print(f"Passenger {name} is at {seat}")
```

What program does:
creates bus passenger list file, then reads it in two ways.

Which code part demonstrates:
open("bus_passengers.txt", "r") with "r" mode and reader.read() / for line in reader demonstrates reading concept.

How it works: 
with open() opens file safely. First read gets full text at once. Second loop reads line by line, splits data, processes it. File auto closes after with block.

WRITING TEXT FILE:

Writing text files means;
--> Creating a new file or overwriting data into an existing .txt file using Python. 
--> We use open(file, "w") for write mode which creates new or clears old data. 
--> Use open(file, "a") for append mode which adds data at end without deleting. 
--> We write using write() method and close file with with.

```python
# Writing Text Files:

# w mode - creates new file / overwrites
with open("coffee_orders.txt", "w") as writer:
    writer.write("Order 1: Ananya - Latte\n")
    writer.write("Order 2: Vivek - Espresso\n")

print("After w mode:")
with open("coffee_orders.txt", "r") as r:
    print(r.read())

# a mode - adds without deleting
with open("coffee_orders.txt", "a") as writer:
    writer.write("Order 3: Zoya - Cappuccino\n")

print("After a mode:")
with open("coffee_orders.txt", "r") as r:
    print(r.read())
```

What program does:
Creates coffee orders file and adds orders.

Which code part demonstrates:
open(..., "w") with write() shows writing, open(..., "a") shows appending.

How it works:
First block with "w" creates file and writes 2 orders. Second read shows it. Third block with "a" adds 3rd order at end without clearing first 2. Final read shows all 3 orders saved.

CSV FILES:

-->CSV means Comma Separated Values. 
--> It is like Excel sheet but in text format, each line is a row and values separated by comma. 
--> Python has csv module to read/write easily. 
--> Better than txt because it handles rows and columns properly. 
--> We use csv.reader to read and csv.writer to write.

```python
# CSV Files:

import csv

# Writing CSV
with open("library_books.csv", "w", newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["book_id", "title", "issued_to"])
    writer.writerow([101, "Python Basics", "Nisha"])
    writer.writerow([102, "Data Science", "Rohan"])

print("CSV written. Now reading:")

# Reading CSV
with open("library_books.csv", "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(f"Book {row['book_id']} - {row['title']} issued to {row['issued_to']}")
```

What program does:
Creates library books CSV and reads it back.

Which code part demonstrates:
csv.writer with writerow() shows writing CSV, csv.DictReader shows reading concept.

How it works:
Writer creates file with header and 2 rows, comma separated. DictReader reads header as keys, each row as dictionary so we can access by column name row['title']. This is how real Excel-like data is handled in file handling.

JSON FILES:

--> JSON is JavaScript Object Notation. 
--> It stores data as key-value pairs like dictionary, easy for Python to use. 
--> We use json module. json.dump() writes dict/list to file, json.load() reads file back to dict. 
--> Best for config, APIs, structured data. Unlike CSV, it keeps nested data.

```python
# JSON Files:

import json

# Data to write
movie_data = {
    "theatre": "PVR Hyderabad",
    "movie": "Kalki",
    "seats": ["A1", "A2", "A3"],
    "price": 250
}

# Writing JSON
with open("ticket.json", "w") as file:
    json.dump(movie_data, file, indent=4)

print("JSON written. Now reading:")

# Reading JSON
with open("ticket.json", "r") as file:
    loaded = json.load(file)
    print(f"Movie: {loaded['movie']} at {loaded['theatre']}")
    print(f"Seats booked: {', '.join(loaded['seats'])}")
    print(f"Total: Rs.{loaded['price'] * len(loaded['seats'])}")
```

What program does:
Creates movie ticket JSON file and reads it.

Which code part demonstrates:
json.dump(movie_data, file) shows writing, json.load(file) shows reading.

How it works:
Dump converts Python dict to JSON text with indentation. Load converts JSON file back to Python dict, so we can directly use keys like loaded['movie'] and nested list loaded['seats']. CSV can't handle list inside easily, JSON can.

PICKLE FILES:

--> Pickle is Python-only file format that saves any Python object as it is, like list, dict, class object. 
--> We use pickle module. pickle.dump() saves object in binary mode wb, pickle.load() reads back in rb mode. 
--> Unlike JSON/CSV, it keeps Python data types and objects exactly same. File not readable by humans.

```python
# Pickle Files:

import pickle

class GameSave:
    def __init__(self, player, level, coins):
        self.player = player
        self.level = level
        self.coins = coins

# Create object
save1 = GameSave("Ishaan", 5, 1200)

# Writing Pickle (binary)
with open("game.dat", "wb") as file:
    pickle.dump(save1, file)

print("Game saved. Now loading:")

# Reading Pickle
with open("game.dat", "rb") as file:
    loaded_save = pickle.load(file)
    print(f"Player: {loaded_save.player}")
    print(f"Level: {loaded_save.level}")
    print(f"Coins: {loaded_save.coins}")
```

What program does:
saves game object to file and loads back.

Which code part demonstrates: 
pickle.dump(save1, file) with "wb" and pickle.load(file) with "rb" shows pickle concept.

How it works: Dump converts custom class object GameSave directly into binary and saves. Load brings back exact same object with all attributes player, level, coins intact. JSON cannot save class object directly, pickle can.

CONTEXT MANAGER(with):

--> Context Manager with with keyword auto handles setup and cleanup. 
--> In file handling, you open file and must close it. 
--> If you forget close or error happens, file stays open and data may corrupt. with open() automatically closes file even if error occurs. 
--> It uses *enter* and *exit* methods internally.

```python
# Context Manager (with):

# Without with - risky
# f = open("exam_result.txt", "w")
# f.write("data")
# if error -> close() may never happen

# With with - safe
with open("exam_result.txt", "w") as report:
    report.write("Student: Kavya\n")
    report.write("Score: 92/100\n")
    print(f"File closed inside with? {report.closed}")

print(f"File closed after with? {report.closed}")

print("\nReading with:")
with open("exam_result.txt", "r") as report:
    print(report.read())
```

What program does:
Writes and reads exam result using with.

Which code part demonstrates:
with open(...) as report: is context manager.

How it works:
Inside with block, file is open closed = False. As soon as block ends, Python auto calls *exit* and closes file closed = True, even if error happened. So no need to write close() manually, safe and clean.

PATHLIB:

--> Pathlib is modern way to handle file paths in Python. 
--> Instead of using string like "folder/file.txt" and os.path, we use Path object. 
--> It is object-oriented, works on Windows and Linux both without changing slash. 
--> It can check if file exists, create folders, read/write directly without open.

```python
# Pathlib:

from pathlib import Path

# Create Path object
base = Path("college_notes")
base.mkdir(exist_ok=True)  # creates folder if not exists

notes_file = base / "syllabus.txt"  # / joins path smartly

# Write directly without open()
notes_file.write_text("Subject: AI\nUnit 1: Done\n")

print(f"File name: {notes_file.name}")
print(f"Parent: {notes_file.parent}")
print(f"Exists? {notes_file.exists()}")
print(f"Content: {notes_file.read_text()}")

# List files in folder
print("\nFiles in college_notes:")
for p in base.iterdir():
    print(f" - {p.name}")
```

What program does:
Creates folder and file using pathlib and reads info.

Which code part demonstrates:
Path(), base / "syllabus.txt", write_text(), read_text(), exists(), iterdir() show pathlib concept.

How it works:
Path object represents path, not string. / operator joins paths correctly for any OS. write_text/read_text does open and close internally. Methods like name, parent, exists() give file info without extra modules.

OS:

--> os module lets Python talk to operating system for file handling tasks that open() can't do. 
--> Like checking if file exists, deleting file, renaming, creating folders, listing files, getting current directory. 
--> It is lower-level control for managing files on disk.

```python
# os in File Handling:

import os

# Create folder
os.makedirs("canteen_bills", exist_ok=True)
print(f"Current dir: {os.getcwd()}")

# Create file using normal write
with open("canteen_bills/bill1.txt", "w") as f:
    f.write("Samosa - Rs.20")

print(f"Exists? {os.path.exists('canteen_bills/bill1.txt')}")
print(f"Files: {os.listdir('canteen_bills')}")

# Rename file
os.rename("canteen_bills/bill1.txt", "canteen_bills/bill_final.txt")
print(f"After rename: {os.listdir('canteen_bills')}")

# Delete file and folder
os.remove("canteen_bills/bill_final.txt")
os.rmdir("canteen_bills")
print(f"After delete, folder exists? {os.path.exists('canteen_bills')}")
```

What program does:
Creates, checks, renames, deletes files/folders using os.

Which code part demonstrates:
os.makedirs, os.path.exists, os.listdir, os.rename, os.remove, os.rmdir shows os concept.

How it works:
makedirs creates folder, exists checks file present, listdir lists files inside, rename changes name without opening, remove/rmdir deletes file and folder from OS. This is OS-level file handling, not content handling.

SHUTIL:

--> shutil means Shell Utilities. 
--> It is higher-level than os. 
--> While os can rename/delete one file, shutil can copy full file content, copy whole folder with all files inside, move files, and zip folders. 
--> It is used for bulk file operations.

```python
# shutil:

import shutil
import os

# Setup source
os.makedirs("project_v1", exist_ok=True)
with open("project_v1/app.py", "w") as f:
    f.write("print('v1 app')")

# 1. Copy single file
shutil.copy("project_v1/app.py", "app_backup.py")
print(f"File copied: {os.path.exists('app_backup.py')}")

# 2. Copy full folder
shutil.copytree("project_v1", "project_v2")
print(f"Folder copied: {os.listdir('project_v2')}")

# 3. Make zip of folder
shutil.make_archive("project_backup", "zip", "project_v1")
print(f"Zip created: {os.path.exists('project_backup.zip')}")

# Cleanup for demo
shutil.rmtree("project_v1")
shutil.rmtree("project_v2")
os.remove("app_backup.py")
os.remove("project_backup.zip")
```

What program does:
Copies file, copies folder, creates zip.

Which code part demonstrates:
shutil.copy(), shutil.copytree(), shutil.make_archive() shows shutil concept.

How it works:
copy copies content from one file to another. copytree copies entire directory with all files inside, like duplicate folder. make_archive zips folder into zip file. os alone cannot do full folder copy or zip, shutil can.

FILE VALIDATION:

--> File Validation means checking file before using it, so code doesn't crash. 
--> We check if file exists, if extension is correct like .jpg, if file size is not too big, if file is not empty. 
--> It is important in real projects when user uploads file, we must validate first.

```python
# File Validation:

import os

# Create test file
with open("resume.pdf", "w") as f:
    f.write("My Resume Content")

def validate_upload(file_path):
    print(f"Validating: {file_path}")
    
    # 1. Exists check
    if not os.path.exists(file_path):
        return "Fail: File not found"
    
    # 2. Extension check
    if not file_path.endswith((".pdf", ".docx")):
        return "Fail: Only pdf/docx allowed"
    
    # 3. Size check (max 5 MB)
    size = os.path.getsize(file_path)
    if size > 5 * 1024 * 1024:
        return "Fail: File too large"
    
    # 4. Empty check
    if size == 0:
        return "Fail: File empty"
    
    return f"Pass: Valid file ({size} bytes)"

print(validate_upload("resume.pdf"))
print(validate_upload("photo.jpg"))
print(validate_upload("missing.txt"))

# cleanup
os.remove("resume.pdf")
```

What program does:
validates uploaded file with 4 checks.

Which code part demonstrates:
os.path.exists, endswith(), os.path.getsize(), size==0 shows file validation concept.

How it works:
Function first checks existence, then checks extension allowed only pdf/docx, then checks size less than 5MB, then checks not empty. Only if all pass returns Pass. This prevents crash and wrong file upload in real apps.

DIRECTORY OPERATIONS:

--> Directory Operations means managing folders, not files. 
--> Main operations are create folder, delete folder, list what's inside, check if folder exists, change current working directory, and navigate inside folders. We use os and pathlib for this.

```python
# Directory Operations:

import os
import shutil

print(f"Start in: {os.getcwd()}")

# 1. Create directories
os.makedirs("college/sem1", exist_ok=True)
print("Created: college/sem1")

# 2. Check exists
print(f"college exists? {os.path.exists('college')}")

# 3. List content
print(f"Inside college: {os.listdir('college')}")

# 4. Create a file inside to demo
with open("college/sem1/math.txt", "w") as f:
    f.write("Algebra")

# 5. Walk through all folders (like search)
print("\nWalking all folders:")
for root, dirs, files in os.walk("college"):
    print(f"Folder: {root} -> Files: {files}")

# 6. Delete everything
shutil.rmtree("college")
print(f"\nAfter delete, exists? {os.path.exists('college')}")
```

What program does:
Creates, lists, walks and deletes directories.

Which code part demonstrates:
os.makedirs, os.path.exists, os.listdir, os.walk, shutil.rmtree shows directory operations.

How it works:
makedirs creates nested folders college/sem1. listdir shows content of folder. os.walk goes through each folder and subfolder and tells root and files, useful for searching. rmtree deletes folder with all its content inside.
