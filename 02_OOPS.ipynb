OBJECT-ORIENTED-PROGRAMMING [OOPS]:

CLASS:

Class is a blueprint or template to create objects. 

It is not real data, but a design that defines what attributes (variables) and behaviors (methods) an object will have.

Instead of writing same code again for similar things, we create one class and use it to create many objects.

```python
# Employee Class:

class Employee:
    def __init__(self, name, emp_id, salary):
        self.name = name
        self.emp_id = emp_id
        self.salary = salary
    
    def get_bonus(self):
        return self.salary * 0.10
    
    def display(self):
        print(f"ID: {self.emp_id}, Name: {self.name}, Salary: {self.salary}, Bonus: {self.get_bonus()}")

# Creating objects from class blueprint
emp1 = Employee("Madhuri", 2749, 50000)
emp2 = Employee("Ravi", 2750, 60000)

print("Employee Details from Class:")
emp1.display()
emp2.display()

print(f"\nemp1 is object of Employee class? {isinstance(emp1, Employee)}")
```

Defines Employee class with name, id, salary and methods to calculate bonus and display. 
Then creates two different employee objects from same class.

Which part demonstrates concept:
class Employee: is the blueprint. *init* is constructor that sets attributes for each object. emp1 = Employee(...) creates real object from class.

How it works: 
Class stores common structure. When we call Employee("Madhuri",...), Python creates new object in memory, *init* assigns values to that object. Each object has its own data but shares same methods. This is OOPS reusability.

OBJECT:

Object is a real instance created from a class. 
We can say class is a blueprint, object is the actual thing built from it.
An object has two things:
*Attributes - data it holds.
*Methods - actions it can do.

```python
# Product as Object

class Product:
    def __init__(self, title, sku_code, stock_count):
        self.title = title
        self.sku_code = sku_code
        self.stock_count = stock_count
    
    def is_available(self):
        return self.stock_count > 0
    
    def show_info(self):
        status = "In Stock" if self.is_available() else "Out of Stock"
        print(f"Product: {self.title} | SKU: {self.sku_code} | Qty: {self.stock_count} | {status}")

# Creating 2 different OBJECTS from same class
item_a = Product("Wireless Mouse", "SKU-101", 15)
item_b = Product("Mechanical Keyboard", "SKU-102", 0)

print("Objects created from Product class:")
item_a.show_info()
item_b.show_info()

print(f"\nitem_a object id: {id(item_a)}")
print(f"item_b object id: {id(item_b)}")
print(f"Are they same object? {item_a is item_b}")
```

What program does: 
Class Product defines structure. item_a and item_b are two separate objects created from it with different title, sku and stock.

Which part demonstrates concept: 
item_a = Product(...) and item_b = Product(...) are objects. id(item_a) and id(item_b) are different proving they are two real instances.

How it works: When we call Product(...), Python allocates new memory and *init* fills that object's attributes. So item_a and item_b have same methods show_info() but hold independent data. Changing one object does not affect the other.

METHOD OVERLOADING-(Python Approach):

Method Overloading means same method name doing different work based on number or type of arguments.

```python
# Method Overloading using Default Arguments

class InvoiceGenerator:
    def generate_total(self, base_price, tax_rate=None, discount_rate=None):
        # Case 1: Only base price
        if tax_rate is None and discount_rate is None:
            return base_price
        
        # Case 2: Base price + tax
        elif discount_rate is None:
            return base_price + (base_price * tax_rate / 100)
        
        # Case 3: Base price + tax + discount
        else:
            taxed = base_price + (base_price * tax_rate / 100)
            return taxed - (taxed * discount_rate / 100)

# One method name, three different uses
bill = InvoiceGenerator()

case1 = bill.generate_total(1000)
case2 = bill.generate_total(1000, 18)
case3 = bill.generate_total(1000, 18, 10)

print("Method Overloading:")
print(f"Only price: {case1}")
print(f"Price + 18% tax: {case2}")
print(f"Price + 18% tax - 10% discount: {case3}")
```

What program does:
Single method generate_total handles 1, 2, and 3 arguments to calculate invoice total in three different ways.

Which part demonstrates concept:
def generate_total(self, base_price, tax_rate=None, discount_rate=None) is Python's way of overloading.
Same method name used as generate_total(1000), generate_total(1000, 18), generate_total(1000, 18, 10).

How it works:
Python checks if tax_rate and discount_rate are None. If None, it skips that calculation. So one method adapts its logic based on arguments received, achieving overloading without defining multiple methods with same name.

CONSTRUCTOR(__init__):


Constructor is a special method *init* that automatically runs when we create an object.
Its work is to initialize the object - set initial values to attributes at the time of creation.
We don't need to call it manually, Python calls it.
Without constructor we have to set values separately after creating object. With constructor, object is ready.

```python
# Constructor:

class LibraryMember:
    def __init__(self, full_name, member_code, join_year):
        print(f"Constructor called for {full_name}")
        self.full_name = full_name
        self.member_code = member_code
        self.join_year = join_year
        self.active = True  # default value set by constructor
    
    def get_membership(self):
        print(f"Member: {self.full_name} | Code: {self.member_code} | Year: {self.join_year} | Active: {self.active}")

# Objects are initialized at creation time by __init__
reader1 = LibraryMember("Kiran Reddy", "LIB-2001", 2023)
reader2 = LibraryMember("Sneha Rao", "LIB-2002", 2024)

print("\nAfter object creation:")
reader1.get_membership()
reader2.get_membership()
```

What program does:
Creates LibraryMember objects where name, code, year and active status are set automatically when object is born.

Which part demonstrates concept:
def *init*(self, full_name, member_code, join_year): is the constructor. The print inside it proves it runs automatically when LibraryMember(...) is called, before get_membership().

How it works:
When we write reader1 = LibraryMember(...), Python creates empty object, then calls *init* with given arguments. self.full_name = full_name assigns values to that specific object. This ensures every object starts with valid initial data without extra method calls.

INSTANCE VARIABLES:

--> Instance variables are variables that belong to a specific object, not to the whole class.
--> Each object has its own copy.They are created using self.variable_name inside *init* or other methods. 
--> If we change value for one object, other objects are not affected.

```python
# Instance Variables:

class BankHolder:
    def __init__(self, holder_name, acc_number, balance_amt):
        # These are instance variables - different for each object
        self.holder_name = holder_name
        self.acc_number = acc_number
        self.balance_amt = balance_amt
    
    def deposit(self, add_amt):
        self.balance_amt += add_amt
    
    def show(self):
        print(f"Name: {self.holder_name} | Acc: {self.acc_number} | Balance: {self.balance_amt}")

# Two objects with different instance variable values
client1 = BankHolder("Vikram Patel", "ACC-9001", 25000)
client2 = BankHolder("Anjali Singh", "ACC-9002", 40000)

print("Before deposit:")
client1.show()
client2.show()

client1.deposit(5000)  # changes only client1's instance variable

print("\nAfter client1 deposits 5000:")
client1.show()
client2.show()

print(f"\nclient1 data: {client1.__dict__}")
print(f"client2 data: {client2.__dict__}")
```

What program does: 
Creates two bank holders where each holds personal name, account number, balance. Deposit to one does not affect other.

Which part demonstrates concept: 
self.holder_name, self.acc_number, self.balance_amt are instance variables. client1.*dict* and client2.*dict* show each object stores separate values.

How it works: When object is created, self points to that particular object, so self.balance_amt becomes that object's own memory. Changing client1.balance_amt updates only memory of client1, client2 remains same. This is independence of instance variables.

CLASS VARIABLES:

Class variable is a variable that belongs to the class itself, not to any one object. It is shared by all objects of that class.

```python
# Class Variables

class UniversityStudent:
    # Class variables - shared by all objects
    university_name = "KL University"
    total_admissions = 0

    def __init__(self, student_name, dept_name):
        self.student_name = student_name
        self.dept_name = dept_name
        UniversityStudent.total_admissions += 1  # class variable updated

    def display_details(self):
        print(f"Student: {self.student_name} | Dept: {self.dept_name} | University: {UniversityStudent.university_name}")

# Creating objects
learner1 = UniversityStudent("Harsha Vardhan", "CSE")
learner2 = UniversityStudent("Divya Sri", "ECE")
learner3 = UniversityStudent("Tarun Kumar", "EEE")

print("Student Details:")
learner1.display_details()
learner2.display_details()
learner3.display_details()

print(f"\nTotal admissions (via class): {UniversityStudent.total_admissions}")
print(f"Total admissions (via learner1): {learner1.total_admissions}")

# Changing class variable affects all
UniversityStudent.university_name = "KLEF Deemed University"

print("\nAfter university name change:")
learner1.display_details()
learner2.display_details()
```

What program does:
Tracks university students where university_name and total_admissions are shared. Counts how many students admitted.

Which part demonstrates concept: 
university_name = "KL University" and total_admissions = 0 defined inside class but outside *init* are class variables. Accessed by UniversityStudent.university_name and UniversityStudent.total_admissions.

How it works: Class variables are stored in class memory, not object memory. All objects refer to same location for university_name. When we do UniversityStudent.total_admissions += 1 in constructor, every new object increments same shared counter. When we change UniversityStudent.university_name, all existing objects immediately see new name because they share it.

INSTANCE METHODS:

Instance method is a method that belongs to an object and works on instance variables. It always takes self as first parameter.

```python
# Instance Methods: 

class CourierPackage:
    def __init__(self, tracking_id, weight_kg, destination_city):
        self.tracking_id = tracking_id
        self.weight_kg = weight_kg
        self.destination_city = destination_city
        self.shipping_cost = 0

    # Instance method - calculates cost for THIS object
    def calculate_cost(self):
        self.shipping_cost = self.weight_kg * 50
        return self.shipping_cost

    # Instance method - uses self data to display
    def update_city(self, new_city):
        self.destination_city = new_city
        print(f"Package {self.tracking_id} rerouted to {self.destination_city}")

    def show_status(self):
        print(f"ID: {self.tracking_id} | Weight: {self.weight_kg}kg | To: {self.destination_city} | Cost: Rs.{self.shipping_cost}")

# Objects
parcel1 = CourierPackage("COUR-301", 2.5, "Hyderabad")
parcel2 = CourierPackage("COUR-302", 5.0, "Chennai")

# Calling instance methods
parcel1.calculate_cost()
parcel2.calculate_cost()

print("Before city update:")
parcel1.show_status()
parcel2.show_status()

parcel1.update_city("Bangalore")

print("\nAfter city update:")
parcel1.show_status()
parcel2.show_status()
```

What program does: 
Two courier packages calculate their own shipping cost and one package updates its destination city.

Which part demonstrates concept: 
def calculate_cost(self):, def update_city(self, new_city):, def show_status(self): are instance methods. They all take self and use self.weight_kg, self.destination_city.

How it works: 
When we call parcel1.calculate_cost(), Python passes parcel1 as self. So self.weight_kg becomes parcel1.weight_kg. Same method when called by parcel2 works on parcel2's data. parcel1.update_city() changed only parcel1's city, proving instance methods operate on individual object data.

STATIC METHOD:

--> Static method is a method that belongs to the class, not to any object. 
--> It does not use self or cls.We define it with @staticmethod decorator. 
--> It cannot access instance variables or class variables directly. 
--> It is like a normal function kept inside class for logical grouping.

```python
# Static Methods:

class LoanApplication:
    bank_name = "SBI Finance"

    def __init__(self, applicant_name, cibil_score, loan_amount):
        self.applicant_name = applicant_name
        self.cibil_score = cibil_score
        self.loan_amount = loan_amount

    @staticmethod
    def is_valid_cibil(score):
        # Utility function - doesn't need self or cls
        return 300 <= score <= 900 and score >= 700

    @staticmethod
    def calculate_emi(principal, annual_rate, months):
        # Pure calculation - no object data needed
        monthly_rate = annual_rate / 12 / 100
        emi = (principal * monthly_rate * (1 + monthly_rate)**months) / ((1 + monthly_rate)**months - 1)
        return round(emi, 2)

    def check_status(self):
        if LoanApplication.is_valid_cibil(self.cibil_score):
            print(f"{self.applicant_name}: Eligible for loan")
        else:
            print(f"{self.applicant_name}: Not Eligible")

# No object needed to call static method
print("Static method direct call:")
print(f"Is 750 valid CIBIL? {LoanApplication.is_valid_cibil(750)}")
print(f"EMI for 5L at 10% for 60 months: Rs.{LoanApplication.calculate_emi(500000, 10, 60)}")

# With objects also
print("\nObject level check:")
client_a = LoanApplication("Pavan Kumar", 780, 500000)
client_b = LoanApplication("Lakshmi", 620, 300000)

client_a.check_status()
client_b.check_status()
```

What program does: 
LoanApplication has two utility tools - CIBIL validator and EMI calculator. They work without object data, but status check uses them.

Which part demonstrates concept: 
@staticmethod above is_valid_cibil and calculate_emi makes them static. They don't have self parameter. Called directly as LoanApplication.is_valid_cibil(750) without creating object.

How it works: 
Static method is stored in class namespace but has no link to instance. Python does not pass self or cls to it. That's why it can be called via class name itself. Inside check_status we call static method via class name to validate object's own cibil score, showing how instance methods can use static methods as helpers.

CLASS METHOD:

--> Class method is a method that belongs to the class, not to object. 
--> It takes cls as first parameter instead of self.cls represents the class itself. 
--> Through cls we can access or modify class variables.
--> We define it with @classmethod decorator. It can be called by class name or object name, but it always works on class data.

```python
# Class Methods: 

class EmployeeRecord:
    organization = "Infosys Solutions"
    work_shift = "9 AM - 5 PM"

    def __init__(self, emp_name, emp_id):
        self.emp_name = emp_name
        self.emp_id = emp_id

    @classmethod
    def change_shift(cls, new_shift):
        # cls refers to class, changes class variable for all
        cls.work_shift = new_shift
        print(f"Work shift updated to {cls.work_shift} by class method")

    @classmethod
    def create_intern(cls, intern_name):
        # Factory method using cls to create object
        return cls(intern_name, "INTERN-TEMP")

    def show(self):
        print(f"Name: {self.emp_name} | ID: {self.emp_id} | Org: {EmployeeRecord.organization} | Shift: {EmployeeRecord.work_shift}")

# Normal objects
staff1 = EmployeeRecord("Rohit Sharma", "EMP-101")
staff2 = EmployeeRecord("Meena Kumari", "EMP-102")

print("Before shift change:")
staff1.show()
staff2.show()

# Calling class method via class name
EmployeeRecord.change_shift("10 AM - 6 PM")

print("\nAfter shift change:")
staff1.show()
staff2.show()

# Factory class method
print("\nFactory method:")
intern = EmployeeRecord.create_intern("Ajay Varma")
intern.show()
```

What program does: 
Employee records share organization and shift. Class method changes shift for all at once and creates intern objects via factory method.

Which part demonstrates concept:
@classmethod with def change_shift(cls, new_shift): and def create_intern(cls, intern_name):. Both use cls not self. cls.work_shift = new_shift modifies class variable.

How it works: 
When we call EmployeeRecord.change_shift(...), Python passes class EmployeeRecord as cls. So cls.work_shift changes the single copy of class variable, and all objects see new value. Second class method create_intern uses cls(...) to call constructor and return new object, showing alternative way to create instances using class method.

PROPERTIES(@propert):

--> @property allows us to use a method like an attribute. 
--> It gives getter, setter control without changing how user accesses it.
--> Normally we call method as obj.method(). With @property, we call obj.method without brackets, but internally a method runs.
--> It is used for validation, computed values, and encapsulation. 
--> We can control what value is allowed when setting, and hide internal variable.

```python
# @property: 

class ProductPricing:
    def __init__(self, product_title, price_value):
        self.product_title = product_title
        self._price_value = price_value  # private variable

    @property
    def price_value(self):
        # getter - called when we access .price_value
        print("Getter called")
        return self._price_value

    @price_value.setter
    def price_value(self, new_price):
        # setter - called when we set .price_value = x
        print(f"Setter called with {new_price}")
        if new_price <= 0:
            print("Error: Price must be positive!")
        else:
            self._price_value = new_price

    @property
    def price_with_tax(self):
        # read-only computed property
        return round(self._price_value * 1.18, 2)

# Create object
item1 = ProductPricing("Wireless Mouse", 800)

print(f"Product: {item1.product_title}")
print(f"Price: {item1.price_value}")  # looks like attribute, but method runs
print(f"Price with 18% GST: {item1.price_with_tax}")

print("\nTrying to update price:")
item1.price_value = 1000  # looks like assignment, but setter method runs
print(f"New Price: {item1.price_value}")

print("\nTrying invalid price:")
item1.price_value = -500
print(f"Price after invalid attempt: {item1.price_value}")
```

What program does: 
Controls product price. Getter shows price, setter validates it must be positive, and computed property gives price with GST.

Which part demonstrates concept: 
@property on def price_value(self): makes it getter. @price_value.setter makes setter. @property on price_with_tax makes read-only computed property. _price_value is internal storage.

How it works: 
When we write item1.price_value, Python internally calls getter method. When we write item1.price_value = 1000, setter method runs and validates. This hides _price_value and provides controlled access like attribute but with logic. price_with_tax has no setter, so it is read-only and always calculated from current price.

ENCAPSULATION:

Encapsulation means binding data and methods together in one class and hiding internal details from outside.

In Python we do it using access modifiers:public - normal access varprotected - _var - hint that don't access directlyprivate - *var - cannot access directly from outside, name mangled to _ClassName*var

```python
# Encapsulation:

class SecureAtmAccount:
    def __init__(self, customer_name, initial_pin):
        self.customer_name = customer_name  # public
        self._branch_code = "HYD-001"       # protected
        self.__pin_number = initial_pin     # private
        self.__account_balance = 0          # private

    # Public method to safely access private data
    def add_funds(self, amount):
        if amount > 0:
            self.__account_balance += amount
            print(f"Rs.{amount} added. New balance: {self.__account_balance}")
        else:
            print("Invalid amount")

    def withdraw_funds(self, amount, entered_pin):
        if entered_pin != self.__pin_number:
            print("Wrong PIN! Access denied")
            return
        if amount <= self.__account_balance:
            self.__account_balance -= amount
            print(f"Rs.{amount} withdrawn. Remaining: {self.__account_balance}")
        else:
            print("Insufficient balance")

    def get_balance(self, entered_pin):
        if entered_pin == self.__pin_number:
            return self.__account_balance
        else:
            return "Access denied - Invalid PIN"

# Create object
client = SecureAtmAccount("Suresh Reddy", 1234)

print(f"Public access: {client.customer_name}")
print(f"Protected access: {client._branch_code}")

# Private direct access fails
try:
    print(client.__account_balance)
except AttributeError as e:
    print(f"Private access failed: {e}")

# Correct way via methods
client.add_funds(10000)
client.withdraw_funds(3000, 1234)
client.withdraw_funds(2000, 9999)  # wrong pin

print(f"\nBalance check with correct PIN: {client.get_balance(1234)}")
print(f"Balance check with wrong PIN: {client.get_balance(1111)}")

# Direct tampering not possible
client.__account_balance = 999999  # creates new public var, not private
print(f"\nAfter tampering attempt, real balance: {client.get_balance(1234)}")
print(f"Tampered var: {client.__dict__}")
```

ABSTRACTION:

Abstraction means hiding complex implementation and showing only essential features to user.

```python
# Abstraction:

from abc import ABC, abstractmethod

# Abstract class - defines WHAT to do, not HOW
class PaymentProcessor(ABC):
    
    @abstractmethod
    def make_payment(self, amount):
        pass

    @abstractmethod
    def refund_payment(self, transaction_id):
        pass

# Child class 1 - provides HOW
class UpiProcessor(PaymentProcessor):
    def make_payment(self, amount):
        print(f"UPI: Processing Rs.{amount} via UPI ID scan...")
        return f"UPI-SUCCESS-{amount}"

    def refund_payment(self, transaction_id):
        print(f"UPI: Refunding for {transaction_id} to original UPI account")

# Child class 2 - different HOW
class CreditCardProcessor(PaymentProcessor):
    def make_payment(self, amount):
        print(f"Credit Card: Charging Rs.{amount} with OTP verification...")
        return f"CC-SUCCESS-{amount}"

    def refund_payment(self, transaction_id):
        print(f"Credit Card: Refunding for {transaction_id} to card in 5-7 days")

# User code - only knows abstract interface
def checkout(payment_obj, amount):
    # user doesn't care if UPI or Card, just calls same methods
    txn = payment_obj.make_payment(amount)
    print(f"Transaction done: {txn}")

# Abstract class object creation fails
# obj = PaymentProcessor() # Error

print("Using UPI:")
upi = UpiProcessor()
checkout(upi, 1500)
upi.refund_payment("UPI-SUCCESS-1500")

print("\nUsing Credit Card:")
card = CreditCardProcessor()
checkout(card, 3200)
card.refund_payment("CC-SUCCESS-3200")
```

What program does: 
Defines payment system where abstract class says every payment must have make_payment and refund_payment, but UPI and Credit Card implement differently. Checkout function works with any payment type.

Which part demonstrates concept: 
class PaymentProcessor(ABC) with @abstractmethod methods make_payment and refund_payment is abstraction. No implementation in base class. Child classes UpiProcessor and CreditCardProcessor must implement them.

How it works: Abstract class hides implementation - it only declares method names. User function checkout knows only abstract interface, not internal OTP or UPI scan logic. If a child does not implement abstract methods, Python throws error at object creation. This enforces standard and hides complexity from end user.

INHERITANCE:

--> Inheritance means one class acquires properties and methods of another class.

--> Parent class is base class, child class is derived class. Child gets all public methods and variables of parent, and can also add its own or override parent's methods.
--> It promotes code reuse - no need to write same code again. Child uses super() to call parent constructor or methods.

```python
# Inheritance:

class BaseStaff:
    def __init__(self, full_name, base_salary):
        self.full_name = full_name
        self.base_salary = base_salary

    def show_profile(self):
        print(f"Staff: {self.full_name} | Base Salary: Rs.{self.base_salary}")

    def calculate_bonus(self):
        return self.base_salary * 0.10

class ManagerStaff(BaseStaff):  # Inherits BaseStaff
    def __init__(self, full_name, base_salary, team_size):
        super().__init__(full_name, base_salary)  # call parent __init__
        self.team_size = team_size

    def calculate_bonus(self):  # Method overriding
        return self.base_salary * 0.25 + (self.team_size * 1000)

    def conduct_meeting(self):
        print(f"{self.full_name} is conducting meeting for {self.team_size} members")

# Object of child
mgr = ManagerStaff("Kiran Kumar", 80000, 6)

# Child can use parent methods
mgr.show_profile()

# Child overridden method
bonus = mgr.calculate_bonus()
print(f"Bonus for Manager: Rs.{bonus}")

# Child own method
mgr.conduct_meeting()

print("\nParent object:")
staff = BaseStaff("Anil Kumar", 40000)
staff.show_profile()
print(f"Bonus for Staff: Rs.{staff.calculate_bonus()}")

print(f"\nIs Manager child of BaseStaff? {issubclass(ManagerStaff, BaseStaff)}")
print(f"Is mgr instance of BaseStaff? {isinstance(mgr, BaseStaff)}")
```

What program does: 
BaseStaff has profile and bonus logic. ManagerStaff inherits it, adds team_size and overrides bonus calculation and adds meeting method.

Which part demonstrates concept: 
class ManagerStaff(BaseStaff): shows inheritance. super().*init* calls parent constructor. calculate_bonus in child overrides parent version. Child object mgr can call parent's show_profile().

How it works: 
When ManagerStaff object is created, its *init* first calls parent's *init* to set full_name and base_salary, then sets its own team_size. Inheritance avoids rewriting name/salary logic. Method overriding lets manager have higher bonus logic. isinstance and issubclass prove inheritance relationship.

MULTIPLE INHERITANCE:

--> Multiple Inheritance means one child class inherits from more than one parent class.
--> If both parents have same method, left parent's method is used unless child overrides.

```python
# Multiple Inheritance: 

class ResearchExpert:
    def __init__(self):
        self.research_domain = "Machine Learning"

    def publish_paper(self):
        print(f"Publishing paper in domain: {self.research_domain}")

class CodingExpert:
    def __init__(self):
        self.programming_lang = "Python"

    def build_project(self):
        print(f"Building project using {self.programming_lang}")

class SmartTeachingAssistant(ResearchExpert, CodingExpert):  # Two parents
    def __init__(self, assistant_name):
        # Must call both parent inits explicitly
        ResearchExpert.__init__(self)
        CodingExpert.__init__(self)
        self.assistant_name = assistant_name

    def assist_students(self):
        print(f"{self.assistant_name} can guide in {self.research_domain} and {self.programming_lang}")

# Object
ta = SmartTeachingAssistant("Sai Teja")

ta.publish_paper()      # From Parent 1
ta.build_project()      # From Parent 2
ta.assist_students()    # Own method

print(f"\nName: {ta.assistant_name} | Domain: {ta.research_domain} | Lang: {ta.programming_lang}")

# MRO - Method Resolution Order
print("\nMRO Order:")
for cls in SmartTeachingAssistant.__mro__:
    print(f" -> {cls.__name__}")
```

What program does: 
ResearchExpert has research skills, CodingExpert has coding skills. SmartTeachingAssistant inherits both and can do publishing, project building and student assistance.

Which part demonstrates concept: 
class SmartTeachingAssistant(ResearchExpert, CodingExpert): inherits from two classes. ta.publish_paper() comes from first parent, ta.build_project() from second parent, proving multiple inheritance.

How it works: 
Child's *init* calls both parents' *init* to set research_domain and programming_lang. Python MRO defines search order: child -> ResearchExpert -> CodingExpert -> object. When we call a method, Python searches in this order. Child object holds attributes of both parents, achieving combined functionality.

MULTILEVEL INHERITANCE:

--> Multilevel Inheritance means chain of inheritance: Grandparent -> Parent -> Child.
--> Child inherits from Parent, Parent inherits from Grandparent. 
--> So child gets properties of both Parent and Grandparent.
--> Each level adds new features to previous level. We use super() to call next level in chain.

```python
# Multilevel Inheritance:

class PersonInfo:  # Level 1 - Grandparent
    def __init__(self, person_name, age_years):
        self.person_name = person_name
        self.age_years = age_years

    def show_person(self):
        print(f"Person: {self.person_name}, Age: {self.age_years}")

class GraduateInfo(PersonInfo):  # Level 2 - Parent
    def __init__(self, person_name, age_years, bachelor_degree):
        super().__init__(person_name, age_years)
        self.bachelor_degree = bachelor_degree

    def show_graduate(self):
        self.show_person()
        print(f"Graduation: {self.bachelor_degree}")

class PostGraduateInfo(GraduateInfo):  # Level 3 - Child
    def __init__(self, person_name, age_years, bachelor_degree, master_degree):
        super().__init__(person_name, age_years, bachelor_degree)
        self.master_degree = master_degree

    def show_postgraduate(self):
        self.show_graduate()
        print(f"Masters: {self.master_degree}")
        print(f"Full Qualification: {self.bachelor_degree} -> {self.master_degree}")

# Child object has access to all levels
student = PostGraduateInfo("Vamsi Krishna", 24, "B.Tech CSE", "M.Tech AI")

print("Level 1 Method (from PersonInfo):")
student.show_person()

print("\nLevel 2 Method (from GraduateInfo):")
student.show_graduate()

print("\nLevel 3 Method (own):")
student.show_postgraduate()

print("\nMRO Chain:")
print(PostGraduateInfo.__mro__)
```

What program does: 
Three level chain - PersonInfo has basic details, GraduateInfo adds bachelor degree, PostGraduateInfo adds master degree. Child object can show all levels.

Which part demonstrates concept: 
class GraduateInfo(PersonInfo): is level 2 inheriting level 1, class PostGraduateInfo(GraduateInfo): is level 3 inheriting level 2. This creates multilevel chain PersonInfo -> GraduateInfo -> PostGraduateInfo.

How it works: 
When PostGraduateInfo is created, its *init* calls super().*init* which goes to GraduateInfo, which again calls super().*init* to PersonInfo. So attributes person_name, age_years, bachelor_degree, master_degree are all set in one chain. Child can call grandparent method show_person() directly because inheritance is transitive. MRO shows order: Child -> Parent -> Grandparent -> object.

HIERARCHICAL INHERITANCE:

--> Hierarchical Inheritance means one parent class and multiple child classes inherit from it.
--> Parent holds common features. Each child gets those common features and adds its own unique behavior. 
--> All children are at same level, sharing one base.

```python
# Hierarchical Inheritance:

class PlatformUser:  # Single Parent
    def __init__(self, user_id, email_address):
        self.user_id = user_id
        self.email_address = email_address

    def login_platform(self):
        print(f"User {self.user_id} logged in with {self.email_address}")

    def logout_platform(self):
        print(f"User {self.user_id} logged out")

# Child 1
class BuyerCustomer(PlatformUser):
    def __init__(self, user_id, email_address, cart_items):
        super().__init__(user_id, email_address)
        self.cart_items = cart_items

    def purchase_items(self):
        print(f"Buyer {self.user_id} purchased {self.cart_items} items")

# Child 2
class SellerMerchant(PlatformUser):
    def __init__(self, user_id, email_address, store_name):
        super().__init__(user_id, email_address)
        self.store_name = store_name

    def add_product(self):
        print(f"Seller {self.user_id} added product to {self.store_name}")

# Child 3
class DeliveryAgent(PlatformUser):
    def __init__(self, user_id, email_address, vehicle_number):
        super().__init__(user_id, email_address)
        self.vehicle_number = vehicle_number

    def deliver_order(self):
        print(f"Agent {self.user_id} delivering via {self.vehicle_number}")

# All children use parent login but have own work
print("Child 1 - Buyer:")
buyer = BuyerCustomer(101, "buyer@gmail.com", 3)
buyer.login_platform()
buyer.purchase_items()
buyer.logout_platform()

print("\nChild 2 - Seller:")
seller = SellerMerchant(202, "seller@shop.com", "RK Electronics")
seller.login_platform()
seller.add_product()

print("\nChild 3 - Delivery Agent:")
agent = DeliveryAgent(303, "agent@logistics.com", "AP-16-AB-1234")
agent.login_platform()
agent.deliver_order()
```

What program does:
PlatformUser is common parent with login/logout. BuyerCustomer, SellerMerchant, DeliveryAgent are three different children inheriting same login but doing different jobs - purchasing, adding product, delivering.

Which part demonstrates concept: 
class BuyerCustomer(PlatformUser):, class SellerMerchant(PlatformUser):, class DeliveryAgent(PlatformUser): - three classes inherit single parent PlatformUser. This is hierarchical inheritance - one to many.

How it works: 
Each child's *init* calls super().*init* to set user_id and email_address from parent, so no repetition. All children can call login_platform() and logout_platform() directly. Parent code written once, reused by many children, each child adds specialization like cart_items, store_name, vehicle_number.

POLYMORPHISM:

--> Polymorphism means one name, many forms. Poly = many, morphism = forms.
--> Same method name behaves differently based on object. 
--> In Python we get it mainly by method overriding - child class gives new implementation for parent method.

```python
# Polymorphism: 

class NotificationService:
    def dispatch_message(self, receiver_contact, message_body):
        raise NotImplementedError("Subclass must implement this")

class SmsNotifier(NotificationService):
    def dispatch_message(self, receiver_contact, message_body):
        print(f"[SMS] to {receiver_contact}: {message_body} -> Sent via SMS Gateway")

class EmailNotifier(NotificationService):
    def dispatch_message(self, receiver_contact, message_body):
        print(f"[EMAIL] to {receiver_contact}: {message_body} -> Sent via SMTP Server")

class PushNotifier(NotificationService):
    def dispatch_message(self, receiver_contact, message_body):
        print(f"[PUSH] to {receiver_contact}: {message_body} -> Sent via Firebase Push")

# Polymorphic function - same method name, different behavior
def send_all_notifications(notifier_list, contact, msg):
    for notifier in notifier_list:
        notifier.dispatch_message(contact, msg)  # same call, many forms

# Different objects, same interface
sms_obj = SmsNotifier()
email_obj = EmailNotifier()
push_obj = PushNotifier()

all_notifiers = [sms_obj, email_obj, push_obj]

send_all_notifications(all_notifiers, "9876543210 / user@mail.com", "Your order is shipped!")

# Also show individual polymorphism
print("\nChecking type-wise:")
for obj in all_notifiers:
    print(f"{obj.__class__.__name__} does: ", end="")
    obj.dispatch_message("user123", "Test")
```

What program does: 
Three notifier classes have same method dispatch_message but different implementation. Function send_all_notifications loops over list and calls same method on each object, getting different output.

Which part demonstrates concept: 
All three classes override dispatch_message. Call notifier.dispatch_message(...) is one name but executes three different forms depending on object - this is polymorphism. Function does not check if object is SMS or Email, it just calls method (duck typing).

How it works: 
Python decides method at runtime based on actual object type, not variable type. notifier_list holds different objects but same interface. When loop runs, for SmsNotifier object it executes SMS logic, for EmailNotifier it executes Email logic. Same method name, many behaviors, code stays clean and extensible - add new notifier without changing send_all_notifications.

METHOD OVERRIDING:

Method Overriding means child class provides its own implementation and its own logic, but with the same parameters in parent class for a method that already exists in parent class.

```python
# Method Overriding: 

class LoanAccount:
    def __init__(self, customer_name, loan_amount):
        self.customer_name = customer_name
        self.loan_amount = loan_amount

    def calculate_interest(self):
        # Base interest 8%
        interest = self.loan_amount * 0.08
        print(f"Base Loan interest for {self.customer_name}: Rs.{interest}")
        return interest

    def show_details(self):
        print(f"Customer: {self.customer_name} | Loan: Rs.{self.loan_amount}")

class HomeLoanAccount(LoanAccount):  # Child overrides method
    def __init__(self, customer_name, loan_amount, property_value):
        super().__init__(customer_name, loan_amount)
        self.property_value = property_value

    def calculate_interest(self):  # Overriding parent method
        # Home loan has lower interest 7% + extra benefit
        base_interest = super().calculate_interest()  # Optionally call parent
        new_interest = self.loan_amount * 0.07
        discount = base_interest - new_interest
        print(f"Home Loan interest for {self.customer_name}: Rs.{new_interest} (You save Rs.{discount})")
        return new_interest

class PersonalLoanAccount(LoanAccount):  # Another child overrides differently
    def calculate_interest(self):
        # Personal loan has higher interest 12%
        new_interest = self.loan_amount * 0.12
        print(f"Personal Loan interest for {self.customer_name}: Rs.{new_interest} (High Risk Loan)")
        return new_interest

# Parent object uses parent method
print("Parent class method:")
base_customer = LoanAccount("Suresh", 500000)
base_customer.calculate_interest()

print("\nChild 1 overridden method:")
home_customer = HomeLoanAccount("Rajesh", 500000, 7000000)
home_customer.show_details()
home_customer.calculate_interest()

print("\nChild 2 overridden method:")
personal_customer = PersonalLoanAccount("Pavan", 500000)
personal_customer.calculate_interest()
```

What program does: 
Parent LoanAccount calculates 8% interest. HomeLoanAccount overrides to 7% with saving message. PersonalLoanAccount overrides to 12%. Same method name gives different results.

Which part demonstrates concept: 
def calculate_interest(self): defined in both parent and child classes with same name but different body. When child object calls it, child version executes. HomeLoanAccount also uses super().calculate_interest() to show parent logic plus its own.

How it works: 
Python looks for method in child class first. If found, it uses child version, ignoring parent. This is runtime decision. Overriding allows specialization - base class gives generic interest, child classes give loan-type specific interest without changing method name, keeping interface same.

METHOD OVERLOADING(Python Approach):

Method Overloading means same class has same method name but with different number or type of parameters, and it behaves differently based on what arguments you pass.

```python
# Method Overloading:

class BillingCalculator:
    def __init__(self, store_branch):
        self.store_branch = store_branch

    # This single method acts as overloaded method
    def calculate_total(self, *item_prices, discount_percent=0, tax_percent=5):
        """
        Works in 3 overloaded forms:
        1. calculate_total(100, 200) -> just sum
        2. calculate_total(100, 200, discount_percent=10) -> with discount
        3. calculate_total(100, 200, discount_percent=10, tax_percent=8) -> discount + tax
        """
        if not item_prices:
            print(f"[{self.store_branch}] No items in bill")
            return 0

        subtotal = sum(item_prices)
        print(f"[{self.store_branch}] Items: {item_prices} | Subtotal: Rs.{subtotal}")

        # Apply discount if given
        discount_amount = subtotal * (discount_percent / 100)
        after_discount = subtotal - discount_amount

        if discount_percent > 0:
            print(f"  Discount {discount_percent}%: -Rs.{discount_amount}")

        # Apply tax
        tax_amount = after_discount * (tax_percent / 100)
        final_total = after_discount + tax_amount

        print(f"  Tax {tax_percent}%: +Rs.{tax_amount}")
        print(f"  Final Total: Rs.{final_total}\n")
        return final_total

# Creating object
billing = BillingCalculator("Mangalagiri Main Branch")

# Form 1: No discount, default tax 5% - Overloading type 1
print("Case 1 - Simple bill:")
billing.calculate_total(150, 250, 300)

# Form 2: With discount - Overloading type 2
print("Case 2 - With discount:")
billing.calculate_total(150, 250, 300, discount_percent=10)

# Form 3: With discount and custom tax - Overloading type 3
print("Case 3 - With discount + custom tax:")
billing.calculate_total(150, 250, 300, discount_percent=10, tax_percent=12)

# Form 4: Single item - still same method works
print("Case 4 - Single item:")
billing.calculate_total(999, discount_percent=5, tax_percent=18)
```

What program does: 
calculate_total calculates bill. Same method handles 1 item, many items, with discount, with tax. No need for 4 different methods.

Which part demonstrates concept: 
Method def calculate_total(self, _item_prices, discount_percent=0, tax_percent=5): uses _item_prices to accept any number of prices and default args for discount/tax. We call same name with different arguments - Case 1 with 3 prices, Case 4 with 1 price, Case 2 with discount. Same method name, many forms.

How it works: 
Python does not allow two methods with same name, last one overwrites. So we make one flexible method. *item_prices collects all positional prices into tuple. discount_percent=0 and tax_percent=5 are optional - if not passed, default used. Inside we check conditions and compute accordingly. This achieves overloading behavior in Pythonic way - one method doing multiple jobs based on input.

COMPOSITION:

Composition means building a class using other classes as parts. Instead of inheriting, we have objects inside object.

```python
# Composition (has-a relationship):

class PowerUnit:
    def __init__(self, battery_capacity):
        self.battery_capacity = battery_capacity
        self.charge_level = 100

    def supply_power(self):
        return f"PowerUnit {self.battery_capacity}mAh supplying power at {self.charge_level}%"

class WifiModule:
    def __init__(self, ssid_name):
        self.ssid_name = ssid_name

    def connect_network(self):
        return f"WifiModule connected to {self.ssid_name}"

class SecurityCamera:
    def __init__(self, camera_id, resolution_mp):
        self.camera_id = camera_id
        self.resolution_mp = resolution_mp

    def start_recording(self):
        return f"Camera {self.camera_id} recording at {self.resolution_mp}MP"

# Main class COMPOSED of above 3 classes
class SmartHomeHub:
    def __init__(self, hub_name, battery_cap, wifi_ssid, cam_id, cam_res):
        self.hub_name = hub_name
        # Composition: Creating part objects inside container
        self.power_component = PowerUnit(battery_cap)
        self.wifi_component = WifiModule(wifi_ssid)
        self.camera_component = SecurityCamera(cam_id, cam_res)

    def activate_home(self):
        print(f"--- {self.hub_name} Activating ---")
        print(self.power_component.supply_power())
        print(self.wifi_component.connect_network())
        print(self.camera_component.start_recording())
        print(f"{self.hub_name} is fully active!\n")

    def show_composition(self):
        print(f"Hub {self.hub_name} HAS-A:")
        print(f" -> {self.power_component.__class__.__name__}")
        print(f" -> {self.wifi_component.__class__.__name__}")
        print(f" -> {self.camera_component.__class__.__name__}")

# Using composition
my_home = SmartHomeHub(
    hub_name="Mangalagiri Smart Hub",
    battery_cap=10000,
    wifi_ssid="HomeNet_5G",
    cam_id="CAM-01",
    cam_res=12
)

my_home.show_composition()
print()
my_home.activate_home()

# Another hub with different parts - shows reusability
office_hub = SmartHomeHub("Office Hub", 20000, "Office_Wifi", "CAM-02", 24)
office_hub.activate_home()
```

What program does: 
PowerUnit, WifiModule, SecurityCamera are independent part classes. SmartHomeHub creates their objects inside its *init* and uses them in activate_home() to run home.

Which part demonstrates concept: 
Lines self.power_component = PowerUnit(battery_cap) , self.wifi_component = WifiModule(wifi_ssid) , self.camera_component = SecurityCamera(...) - these show composition. SmartHomeHub has-a PowerUnit, has-a WifiModule, has-a SecurityCamera. It does not inherit them.

How it works: 
Container class SmartHomeHub owns the lifecycle of part objects - they are created when hub is created. It delegates work to parts: self.power_component.supply_power() etc. This is loose coupling - we can change battery capacity or wifi without changing hub logic, and we can reuse PowerUnit in other devices like Robot or Drone. Composition is more flexible than inheritance for part-whole relations.

AGGREGATION:

--> In Aggregation, container uses part objects, but does NOT own their lifecycle. 
--> Part objects exist independently outside container. 
--> If container is destroyed, parts still exist.Example: 
--> Team has-a Player, but Player exists even if Team is deleted.Difference: Composition -> parts created inside container. Aggregation -> parts created outside and passed inside.

```python
# Aggregation: 

class MusicTrack:
    def __init__(self, track_title, artist_name, duration_sec):
        self.track_title = track_title
        self.artist_name = artist_name
        self.duration_sec = duration_sec

    def play_track(self):
        return f"Playing '{self.track_title}' by {self.artist_name} [{self.duration_sec}s]"

# Container class that aggregates tracks - it does NOT create tracks
class PlayList:
    def __init__(self, playlist_name, creator_name):
        self.playlist_name = playlist_name
        self.creator_name = creator_name
        self.track_collection = [] # Will hold independent track objects

    def add_track_to_playlist(self, track_obj):
        # Aggregation: adding existing object, not creating new inside
        self.track_collection.append(track_obj)
        print(f"Added '{track_obj.track_title}' to '{self.playlist_name}'")

    def show_all_tracks(self):
        print(f"\nPlaylist: {self.playlist_name} | Created by: {self.creator_name}")
        total_time = 0
        for idx, track in enumerate(self.track_collection, 1):
            print(f" {idx}. {track.play_track()}")
            total_time += track.duration_sec
        print(f"Total Duration: {total_time} seconds")

# Step 1: Tracks exist independently outside playlist (Aggregation key point)
track_one = MusicTrack("Naa Songs", "DSP", 210)
track_two = MusicTrack("Folk Beat", "Ram Miriyala", 180)
track_three = MusicTrack("Melody Night", "Shreya", 240)

print("Tracks exist independently:")
print(track_one.play_track())

# Step 2: Create playlist and aggregate existing tracks
my_playlist = PlayList("Telugu Hits 2026", "Kiran")
my_playlist.add_track_to_playlist(track_one)
my_playlist.add_track_to_playlist(track_two)
my_playlist.add_track_to_playlist(track_three)

my_playlist.show_all_tracks()

# Step 3: Prove aggregation - tracks still exist if playlist deleted
print("\nDeleting playlist object...")
del my_playlist

print("But tracks still exist after playlist deletion (Aggregation proof):")
print(track_two.play_track())
print(track_three.play_track())
```

What program does: 
MusicTrack objects created first independently. PlayList creates empty list and aggregates those existing tracks via add_track_to_playlist. It shows them and calculates total duration. Even after deleting playlist, tracks still playable.

Which part demonstrates concept: 
def add_track_to_playlist(self, track_obj): takes an already created outside object and stores it. Tracks are NOT created with MusicTrack(...) inside PlayList *init* - they are passed from outside. del my_playlist but track_two.play_track() still works - proves weak ownership.

How it works: 
Aggregation uses reference. Playlist holds references to track objects, does not control their lifecycle. Track exists before playlist and after playlist deletion. This models real world: playlist has-a track, but track independent. Unlike composition where parts die with container, here parts survive.

ASSOCIATION: 

--> Association is a general relationship where one class uses another class. 
--> It is "uses-a" connection.
--> Two classes are independent, they know about each other and collaborate. 
--> No ownership like composition/aggregation.
--> It can be one-way or two-way.  neither contains the other permanently.

```python
# Association:

class PatientProfile:
    def __init__(self, patient_code, full_name, symptom_desc):
        self.patient_code = patient_code
        self.full_name = full_name
        self.symptom_desc = symptom_desc

    def get_patient_info(self):
        return f"Patient {self.patient_code}: {self.full_name} - Symptom: {self.symptom_desc}"

class DoctorProfile:
    def __init__(self, doctor_code, doctor_name, specialization_field):
        self.doctor_code = doctor_code
        self.doctor_name = doctor_name
        self.specialization_field = specialization_field

    def diagnose_patient(self, patient_obj):
        # Association: Doctor USES Patient object, but doesn't own it
        print(f"Dr.{self.doctor_name} [{self.specialization_field}] checking {patient_obj.full_name}")
        print(f" -> Info: {patient_obj.get_patient_info()}")
        print(f" -> Diagnosis done for {patient_obj.patient_code}\n")
        return f"Prescription for {patient_obj.full_name}"

    def get_doctor_info(self):
        return f"Doctor {self.doctor_code}: Dr.{self.doctor_name} ({self.specialization_field})"

# Both objects created independently
doc_1 = DoctorProfile("D-101", "Arjun Reddy", "Cardiology")
doc_2 = DoctorProfile("D-102", "Sneha", "Neurology")

pat_1 = PatientProfile("P-501", "Ravi Kumar", "Chest Pain")
pat_2 = PatientProfile("P-502", "Lakshmi", "Headache")

# Association in action - independent objects collaborating
print("Independent Existence:")
print(doc_1.get_doctor_info())
print(pat_1.get_patient_info())
print("\n--- Association Starts (Doctor uses Patient) ---")

doc_1.diagnose_patient(pat_1) # Cardiology doctor checks chest pain patient
doc_2.diagnose_patient(pat_2) # Neurology doctor checks headache patient

# Same patient can associate with different doctor - proves no ownership
print("--- Same patient associates with another doctor ---")
doc_2.diagnose_patient(pat_1)

print("--- Both still exist independently after association ---")
print(doc_1.get_doctor_info())
print(pat_2.get_patient_info())
```

What program does: 
DoctorProfile and PatientProfile created separately. Doctor's method diagnose_patient takes patient object as parameter and uses its data. One patient can be checked by multiple doctors, showing independent life.

Which part demonstrates concept: 
def diagnose_patient(self, patient_obj): - Doctor class uses Patient class object. No creation of Patient inside Doctor, and no list storing permanently. Just method parameter collaboration. That is association: uses-a.

How it works: 
Objects doc_1 and pat_1 exist independently in memory. When we call doc_1.diagnose_patient(pat_1), Doctor temporarily knows Patient and calls patient_obj.get_patient_info(). After method ends, association ends. Neither owns other, both can exist and associate with other objects freely. This is different from composition where parent creates child, and aggregation where container holds list. Here it's just usage.

MAGIC (Dunder) METHODS:

--> Magic methods also called Dunder methods because they have Double Underscore before and after name like *init*, *str*.
--> They are special methods Python calls automatically in certain situations. 
--> You don't call obj.*str*() directly, you call print(obj) and Python internally calls *str*.
--> They let us customize behavior of our class for built-in operations like +, len(), print(), ==, indexing.

```python
# Magic (Dunder) Methods:

class CartItemBox:
    def __init__(self, product_label, unit_cost, quantity_count):
        self.product_label = product_label
        self.unit_cost = unit_cost
        self.quantity_count = quantity_count
        print(f"__init__ called -> Created {self.product_label}")

    def __str__(self):
        # Called when print(object) is used
        return f"[Cart: {self.product_label} | Cost: {self.unit_cost} x Qty: {self.quantity_count} = {self.unit_cost * self.quantity_count}]"

    def __len__(self):
        # Called when len(object) is used
        return self.quantity_count

    def __add__(self, other_box):
        # Called when object1 + object2 is used
        if isinstance(other_box, CartItemBox):
            total_value = (self.unit_cost * self.quantity_count) + (other_box.unit_cost * other_box.quantity_count)
            return CartItemBox(f"Combined-{self.product_label}+{other_box.product_label}", total_value, 1)
        return NotImplemented

    def __eq__(self, other_box):
        # Called when object1 == object2 is used
        return (self.unit_cost * self.quantity_count) == (other_box.unit_cost * other_box.quantity_count)

# __init__ automatically called here
box_a = CartItemBox("Wireless Mouse", 800, 2)
box_b = CartItemBox("Keyboard", 1600, 1)

# __str__ automatically called here
print("\n__str__ magic:")
print(box_a)
print(box_b)

# __len__ automatically called here
print("\n__len__ magic:")
print(f"len(box_a) = {len(box_a)} items")

# __add__ automatically called here
print("\n__add__ magic:")
box_c = box_a + box_b
print(box_c)

# __eq__ automatically called here
print("\n__eq__ magic:")
print(f"box_a == box_b ? {box_a == box_b}")
box_d = CartItemBox("USB Hub", 800, 2)
print(f"box_a == box_d ? {box_a == box_d} (both 1600 total)")
```

What program does: 
CartItemBox represents product. It uses 4 magic methods to make object behave like built-in type - printable, len-able, addable, comparable.

Which part demonstrates concept: 
*init* runs on creation, *str* runs on print(), *len* runs on len(), *add* runs on +, *eq* runs on == . We never called box_a.*str*() directly, Python did.

How it works: 
Python operator maps to dunder. When you write box_a + box_b, Python translates to box_a.*add*(box_b). When you write len(box_a), it calls box_a.*len*(). By defining these we customize default behavior - here + merges two carts into one combined box, == compares total value not object address, len returns quantity, str gives readable bill format.

SUPER():

--> When child overrides parent method or *init*, we lose parent logic. 
--> super() lets us reuse parent logic without rewriting it.
--> It also solves multiple inheritance problem - it follows MRO (Method Resolution Order) and calls next class in chain automatically.

```python
# super() function:

class StaffMember:
    def __init__(self, staff_id, staff_name, base_salary):
        self.staff_id = staff_id
        self.staff_name = staff_name
        self.base_salary = base_salary

    def show_salary_slip(self):
        return f"Staff {self.staff_name} [{self.staff_id}] | Base Salary: Rs.{self.base_salary}"

    def calculate_final_pay(self):
        return self.base_salary

class ManagerStaff(StaffMember):
    def __init__(self, staff_id, staff_name, base_salary, team_size, bonus_percent):
        # super() calls parent __init__ to avoid rewriting id, name, salary logic
        super().__init__(staff_id, staff_name, base_salary)
        self.team_size = team_size
        self.bonus_percent = bonus_percent

    def show_salary_slip(self):
        # super() gets base slip from parent, then we extend it
        parent_slip = super().show_salary_slip()
        return f"{parent_slip} | Team: {self.team_size} | Bonus: {self.bonus_percent}%"

    def calculate_final_pay(self):
        # super() gets base pay, child adds bonus on top
        base_pay = super().calculate_final_pay()
        bonus_amount = base_pay * (self.bonus_percent / 100)
        final = base_pay + bonus_amount
        print(f"Base: {base_pay} + Bonus: {bonus_amount} = Final: {final}")
        return final

# Parent object
print("Parent class:")
emp_1 = StaffMember("S-100", "Harish", 40000)
print(emp_1.show_salary_slip())
print(f"Final Pay: {emp_1.calculate_final_pay()}\n")

# Child object using super()
print("Child class using super():")
mgr_1 = ManagerStaff("M-201", "Divya", 60000, 12, 20)
print(mgr_1.show_salary_slip())
mgr_1.calculate_final_pay()
```

What program does: 
StaffMember has base salary logic. ManagerStaff inherits it, adds team size and bonus. It uses super() to reuse parent init, slip, and pay calculation.

Which part demonstrates concept: 
super().*init*(staff_id, staff_name, base_salary) in child *init* calls parent init - no need to rewrite self.staff_id = ... again. parent_slip = super().show_salary_slip() and base_pay = super().calculate_final_pay() call parent methods inside overridden child methods to extend them.

How it works: 
super() refers to parent class object. When we call super().*init*, Python goes to MRO, finds StaffMember, executes its init, so child object gets staff_id, name, salary attributes. Later when we override show_salary_slip, using super().show_salary_slip() first gets parent string, then child adds team and bonus info. This avoids code duplication and maintains link - if parent salary format changes, child automatically gets updated version via super().

__str__():

--> *str*() is a dunder method that controls what happens when you do print(object) or str(object).
--> Without *str*, Python prints ugly address like <*main*.
--> MyClass object at 0x...> which is not readable.
--> With *str*, you return a clean, user-friendly string. 
--> It should explain object in human language.
--> It does not print by itself - it must return string. 
--> Python calls it automatically when object needs string form.

```python
# __str__() method: 

class DeliveryOrder:
    def __init__(self, order_code, customer_name, total_amount, delivery_status):
        self.order_code = order_code
        self.customer_name = customer_name
        self.total_amount = total_amount
        self.delivery_status = delivery_status

    def __str__(self):
        # This is called automatically by print() and str()
        return (
            f"--- Delivery Slip ---\n"
            f"Order ID   : {self.order_code}\n"
            f"Customer   : {self.customer_name}\n"
            f"Amount     : Rs.{self.total_amount}\n"
            f"Status     : {self.delivery_status}\n"
            f"---------------------"
        )

# Without __str__, print would be unreadable. With __str__, clean output.

parcel_1 = DeliveryOrder("ORD-7845", "Sai Kumar", 1250, "Out for Delivery")
parcel_2 = DeliveryOrder("ORD-7846", "Anjali", 899, "Delivered")

# __str__ automatically invoked
print("Printing parcel_1:")
print(parcel_1)

print("\nPrinting parcel_2:")
print(parcel_2)

print("\nUsing str() function directly:")
order_summary = str(parcel_1)
print(order_summary)

print("\nIn list, Python uses __str__ for each item:")
order_list = [parcel_1, parcel_2]
for single_order in order_list:
    print(single_order)
    print()
```

What program does: 
DeliveryOrder stores order details. *str* returns formatted delivery slip. When we print objects, readable slip appears instead of memory address.

Which part demonstrates concept: 
def *str*(self): return f"--- Delivery Slip ---\nOrder ID..." This method is the concept. Calls like print(parcel_1) and str(parcel_1) automatically trigger it.

How it works:
Python internally calls parcel_1.*str*() when it needs string version. Our implementation builds human-friendly multi-line string using f-string and returns it. Must return string, not print. This makes debugging and UI display clean. If *str* not defined, Python falls back to default *repr* showing object location.

__repr__():

*repr*() is a dunder method that gives official string representation of object for developers.It is called when you type object name in Jupyter, or when you do repr(object), or when object is inside list and printed.

```python
# __repr__() method: 

class LabExperiment:
    def __init__(self, exp_code, exp_title, trial_count, success_rate):
        self.exp_code = exp_code
        self.exp_title = exp_title
        self.trial_count = trial_count
        self.success_rate = success_rate

    def __repr__(self):
        # Official representation - should look like constructor call
        return f"LabExperiment(exp_code='{self.exp_code}', exp_title='{self.exp_title}', trial_count={self.trial_count}, success_rate={self.success_rate})"

    def __str__(self):
        # User friendly version
        return f"Experiment {self.exp_code}: {self.exp_title} | Success {self.success_rate}%"

# Creating objects
exp_1 = LabExperiment("EXP-A1", "Solar Cell Test", 50, 92.5)
exp_2 = LabExperiment("EXP-B2", "Water Purification", 30, 88.0)

print("1. __str__ called by print():")
print(exp_1)

print("\n2. __repr__ called by repr():")
print(repr(exp_1))

print("\n3. In Jupyter / List, __repr__ is used:")
experiment_batch = [exp_1, exp_2]
print(experiment_batch)

print("\n4. __repr__ is recreatable - copy output to create new object:")
repr_string = repr(exp_1)
print(f"Repr string: {repr_string}")

# Recreating object from repr string - proof it is official
exp_recreated = eval(repr_string)
print(f"Recreated object via eval: {exp_recreated}")
print(f"Same data? {exp_recreated.exp_code == exp_1.exp_code and exp_recreated.success_rate == exp_1.success_rate}")
```

What program does: 
LabExperiment has both *str* for pretty display and *repr* for exact developer log. List printing and Jupyter cell output use *repr*. We show repr string can recreate object via eval.

Which part demonstrates concept: 
def *repr*(self): return f"LabExperiment(exp_code='{self.exp_code}', ...)" - this returns code-like string that looks like constructor. That is core of *repr*.

How it works: 
When you write exp_1 alone in Jupyter or print list, Python calls *repr* on each object, not *str*. Our *repr* returns complete initialization string with all attributes, making debugging easy - you instantly see full state. The eval(repr(obj)) recreating object proves representation is unambiguous and official, which is best practice for *repr*.

__len__():

--> *len*() is a dunder method that defines what len(object) should return for your class.
--> Python's built-in len() does not know how to count your custom objects. 
--> When you define *len*, you tell Python how to measure length.
--> It must return integer >= 0. No arguments allowed except self.

```python
# __len__() method: 

class WarehouseInventory:
    def __init__(self, warehouse_name, location_city):
        self.warehouse_name = warehouse_name
        self.location_city = location_city
        self.stock_items = {}  # item_name: quantity

    def add_stock(self, item_name, qty):
        self.stock_items[item_name] = self.stock_items.get(item_name, 0) + qty
        print(f"Added {qty} x {item_name}")

    def __len__(self):
        # len() should return total distinct items count
        # We can customize - here we return total quantity of all items
        total_quantity = sum(self.stock_items.values())
        return total_quantity

    def __str__(self):
        return f"Warehouse {self.warehouse_name} at {self.location_city} | Stock: {self.stock_items}"

# Create warehouse
godown = WarehouseInventory("Central Godown", "Mangalagiri")

godown.add_stock("Cement Bags", 120)
godown.add_stock("Steel Rods", 80)
godown.add_stock("Paint Boxes", 50)

print("\n" + str(godown))

print("\n__len__ magic in action:")
print(f"len(godown) = {len(godown)} total units")

# Adding more stock, len changes automatically
godown.add_stock("Cement Bags", 30) # add 30 more
print(f"\nAfter adding 30 more cement:")
print(f"len(godown) = {len(godown)} total units")
print(godown)

# len() used in condition - professional use
if len(godown) > 200:
    print("\nStatus: Warehouse is FULL ( >200 units )")
else:
    print("\nStatus: Space Available")
```

What program does: 
WarehouseInventory stores items with quantities in dict. add_stock adds items, *str* shows content, *len* returns total sum of all quantities.

Which part demonstrates concept: 
def *len*(self): total_quantity = sum(self.stock_items.values()); return total_quantity This is *len* logic. Calls len(godown) trigger it.

How it works: 
Python len() internally calls godown.*len*(). We customized to return total stock count, not distinct items. You could also return len(self.stock_items) for distinct count, but we chose total quantity to show customization power. It must return int, so we sum values and return. Now len() works on our custom class like built-in list, and can be used in if conditions.

DATACLASSES (BONUS):

--> Dataclasses are shortcut to create classes that mainly hold data. Normal class needs you to write *init*, *repr*, *eq* manually.
--> You just declare fields like name: str and it creates constructor, printable representation, comparison, etc. automatically.

```python
# DATACLASSES: 

from dataclasses import dataclass, field

@dataclass(order=True) # order=True allows sorting by first field
class ScholarshipApplicant:
    # sort_index is used for ordering, not shown in repr
    sort_index: float = field(init=False, repr=False)
    applicant_id: str
    full_name: str
    department_name: str
    cgpa_score: float
    published_papers: int = 0
    is_selected: bool = False

    def __post_init__(self):
        # Called after auto __init__
        # We set sort_index for automatic sorting by cgpa (descending logic handled later)
        self.sort_index = self.cgpa_score
        # Validation logic can be added here
        if not (0 <= self.cgpa_score <= 10):
            raise ValueError("CGPA must be between 0 and 10")

    def eligibility_status(self):
        if self.cgpa_score >= 8.5 and self.published_papers >= 2:
            return "Eligible for Merit Scholarship"
        return "Not Eligible"

# No manual __init__ / __repr__ / __eq__ needed - dataclass made them

print("Creating objects - auto __init__:")
app_1 = ScholarshipApplicant("SCH-101", "Keerthi", "CSE", 9.2, 3)
app_2 = ScholarshipApplicant("SCH-102", "Manoj", "ECE", 8.1, 1)
app_3 = ScholarshipApplicant("SCH-103", "Ayesha", "CSE", 9.5, 4)

print(app_1)
print(app_2)

print("\nAuto __eq__ - comparison by data:")
app_duplicate = ScholarshipApplicant("SCH-101", "Keerthi", "CSE", 9.2, 3)
print(f"app_1 == app_duplicate ? {app_1 == app_duplicate}")

print("\nAuto order=True - sorting by cgpa:")
applicant_list = [app_1, app_2, app_3]
# Sort descending by cgpa using sort_index
sorted_list = sorted(applicant_list, reverse=True)
for single_app in sorted_list:
    print(f"{single_app.full_name} - {single_app.cgpa_score} CGPA - {single_app.eligibility_status()}")

print("\nMethod still works with dataclass:")
print(f"{app_1.full_name} status: {app_1.eligibility_status()}")
```

What program does: 
ScholarshipApplicant dataclass holds student scholarship data. We didn't write *init*, *repr*, *eq*. Decorator made them. We added *post_init* for validation and sort_index setup, and eligibility_status() custom logic.

Which part demonstrates concept:
@dataclass(order=True) and field declarations applicant_id: str etc. show dataclass concept. No manual constructor, yet ScholarshipApplicant("SCH-101", "Keerthi", ...) works. Printing object gives nice representation automatically. Comparison app_1 == app_duplicate works by value, not address.

How it works: 
Decorator reads type hints and generates *init*(self, applicant_id, full_name, ...), *repr* returning all fields, *eq* comparing all fields, and with order=True, comparison methods for sorting. field(repr=False, init=False) hides sort_index from init and print, used only for sorting. *post_init* runs after auto init, good for validation. This cuts 20-30 lines boilerplate to 5 lines.

Singleton Pattern (Bonus):

--> Singleton Pattern ensures a class has only ONE object in whole program, and gives global access to it.
--> If you try to create new object second time, it returns same first object, not new one.

```python
# Singleton Pattern: 

class UniversityAppConfig:
    _single_instance = None  # Holds the only instance
    _is_initialized = False  # To avoid re-running __init__

    def __new__(cls, *args, **kwargs):
        # This is called BEFORE __init__, controls object creation
        if cls._single_instance is None:
            print("__new__: No instance found, creating first and only instance")
            cls._single_instance = super().__new__(cls)
        else:
            print("__new__: Instance already exists, returning same instance")
        return cls._single_instance

    def __init__(self, app_name="SRM University App", version="1.0", maintenance_mode=False):
        # Prevent re-initialization on second object creation
        if not self._is_initialized:
            print("__init__: Initializing config for first time")
            self.app_name = app_name
            self.version = version
            self.maintenance_mode = maintenance_mode
            self.__class__._is_initialized = True
        else:
            print("__init__: Already initialized, skipping re-init")

    def show_config(self):
        return f"App: {self.app_name} | Ver: {self.version} | Maintenance: {self.maintenance_mode}"

# Trying to create multiple objects
print("--- Creating config_1 ---")
config_1 = UniversityAppConfig("MyPortal", "2.5", False)
print(config_1.show_config())
print(f"ID of config_1: {id(config_1)}\n")

print("--- Creating config_2 (different values) ---")
config_2 = UniversityAppConfig("FakePortal", "9.9", True)
print(config_2.show_config())
print(f"ID of config_2: {id(config_2)}\n")

print("--- Checking Singleton behavior ---")
print(f"config_1 is config_2 ? {config_1 is config_2}")
print(f"config_1 ID == config_2 ID ? {id(config_1) == id(config_2)}")

print("\n--- Changing via config_1 affects config_2 ---")
config_1.maintenance_mode = True
print(f"config_1: {config_1.show_config()}")
print(f"config_2: {config_2.show_config()} (same because same object)")
```

What program does: 
UniversityAppConfig should have single config for whole app. First creation makes object with values MyPortal 2.5. Second creation with FakePortal 9.9 does not create new object, returns first one, ignores new values.

Which part demonstrates concept: 
*new* method with _single_instance class variable. if _single_instance is None: create else: return same. And _is_initialized flag stops *init* from overwriting first values on second call. id() same for both proves singleton.

How it works: 
Normally obj = Class() calls *new* to allocate memory, then *init* to fill data. By overriding *new*, we control allocation - first time we allocate and store in _single_instance, second time we skip allocation and return stored reference. So config_1 and config_2 point to same memory. Changing attribute via one changes for all, proving only one object exists.
