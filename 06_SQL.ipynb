RELATIONSHIPS:

A relationship in SQL describes how data in one table is connected to data in another table.

The main types of relationships are:

One-to-One (1:1) – One record in one table is related to one record in another table.
One-to-Many (1:M) – One record in one table can be related to many records in another table.
Many-to-Many (M:M) – Many records in one table can be related to many records in another table. This usually requires a third table called a junction table.

A Primary Key and Foreign Key are commonly used to create these relationships.

| Student_ID (Primary Key) | Name  |
| -----------------------: | ----- |
|                      101 | Rahul |
|                      102 | Priya |
|                      103 | Arjun |

| Course_ID (Primary Key) | Course_Name |
| ----------------------: | ----------- |
|                       1 | Python      |
|                       2 | SQL         |

We can create another table called 
ENROLLMENTS:

| Enrollment_ID | Student_ID (Foreign Key) | Course_ID (Foreign Key) |
| ------------: | -----------------------: | ----------------------: |
|             1 |                      101 |                       1 |
|             2 |                      102 |                       2 |
|             3 |                      103 |                       3 |

```python
import sqlite3

# Create a database
connection = sqlite3.connect("college.db")

# Enable foreign key constraints
connection.execute("PRAGMA foreign_keys = ON")

cursor = connection.cursor()

# Create Students table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS Students (
        student_id INTEGER PRIMARY KEY,
        name TEXT NOT NULL
    )
""")

# Create Courses table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS Courses (
        course_id INTEGER PRIMARY KEY,
        course_name TEXT NOT NULL
    )
""")

# Create Enrollments table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS Enrollments (
        student_id INTEGER,
        course_id INTEGER,

        PRIMARY KEY (student_id, course_id),

        FOREIGN KEY (student_id)
            REFERENCES Students(student_id),

        FOREIGN KEY (course_id)
            REFERENCES Courses(course_id)
    )
""")

# Add students
cursor.executemany("""
    INSERT OR IGNORE INTO Students (student_id, name)
    VALUES (?, ?)
""", [
    (101, "Rahul"),
    (102, "Priya"),
    (103, "Arjun")
])

# Add courses
cursor.executemany("""
    INSERT OR IGNORE INTO Courses (course_id, course_name)
    VALUES (?, ?)
""", [
    (1, "Python"),
    (2, "SQL"),
    (3, "Java")
])

# Connect students with courses
cursor.executemany("""
    INSERT OR IGNORE INTO Enrollments (student_id, course_id)
    VALUES (?, ?)
""", [
    (101, 1),  # Rahul -> Python
    (101, 2),  # Rahul -> SQL
    (102, 2),  # Priya -> SQL
    (103, 3)   # Arjun -> Java
])

connection.commit()

# Display student-course relationships
cursor.execute("""
    SELECT Students.name, Courses.course_name
    FROM Enrollments
    JOIN Students
        ON Enrollments.student_id = Students.student_id
    JOIN Courses
        ON Enrollments.course_id = Courses.course_id
""")

print("Student - Course Relationships:")

for student, course in cursor.fetchall():
    print(student, "->", course)

connection.close()
```

The program creates three tables:

Students stores student information.
Courses stores course information.
Enrollments connects students and courses.

ORDERBY:

ORDER BY is a SQL clause used to sort the results of a query.

By default, SQL can return records in an order that is not necessarily the order you want. ORDER BY lets us specify how the results should be arranged.

We can sort data in two main ways:

ASC (Ascending) – smallest to largest, or A to Z. This is the default.
DESC (Descending) – largest to smallest, or Z to A.

```python
import sqlite3

# Connect to the database
connection = sqlite3.connect("students.db")

# Create a cursor
cursor = connection.cursor()

# Create the Students table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS Students (
        student_id INTEGER PRIMARY KEY,
        name TEXT,
        age INTEGER,
        marks INTEGER
    )
""")

# Insert student records
students = [
    (101, "Rahul", 20, 75),
    (102, "Priya", 21, 90),
    (103, "Arjun", 19, 65),
    (104, "Sneha", 22, 85)
]

cursor.executemany("""
    INSERT OR IGNORE INTO Students
    (student_id, name, age, marks)
    VALUES (?, ?, ?, ?)
""", students)

connection.commit()

# Use ORDER BY to sort marks from highest to lowest
cursor.execute("""
    SELECT name, marks
    FROM Students
    ORDER BY marks DESC
""")

# Get the sorted results
results = cursor.fetchall()

print("Students sorted by marks:")

for student in results:
    print(student)

# Close the connection
connection.close()
```

```python
import sqlite3

# Connect to the database
connection = sqlite3.connect("company.db")

# Create a cursor
cursor = connection.cursor()

# Create the Employees table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS Employees (
        employee_id INTEGER PRIMARY KEY,
        name TEXT,
        department TEXT,
        salary INTEGER
    )
""")

# Insert employee records
employees = [
    (101, "Rahul", "IT", 50000),
    (102, "Priya", "HR", 45000),
    (103, "Arjun", "IT", 60000),
    (104, "Sneha", "HR", 55000),
    (105, "Kiran", "IT", 70000),
    (106, "Anil", "Sales", 40000)
]

cursor.executemany("""
    INSERT OR IGNORE INTO Employees
    (employee_id, name, department, salary)
    VALUES (?, ?, ?, ?)
""", employees)

connection.commit()

# GROUP BY and HAVING
cursor.execute("""
    SELECT department, SUM(salary) AS total_salary
    FROM Employees
    GROUP BY department
    HAVING SUM(salary) > 100000
""")

# Get the results
results = cursor.fetchall()

print("Departments with total salary above 100000:")

for department, total_salary in results:
    print(department, "->", total_salary)

# Close the connection
connection.close()
```

In simple words:

HAVING is used to filter groups after GROUP BY. It is commonly used when we need to filter the result of aggregate functions such as COUNT(), SUM(), or AVG().

FULLJOIN:

A FULL JOIN (also called FULL OUTER JOIN) is used to combine two tables and return all records from both tables.

If a record has a matching value in the other table, the information is combined. 
If there is no match, SQL still keeps the record and fills the missing side with NULL.

EXAMPLE:
PROJECTS TABLE:

| Project_ID | Project_Name | Manager_ID |
| ---------: | ------------ | ---------: |
|        201 | Website      |        501 |
|        202 | Mobile App   |        502 |
|        203 | Database     |        503 |

MANAGERS TABLE:

| Manager_ID | Manager_Name |
| ---------: | ------------ |
|        501 | Meera        |
|        502 | Vikram       |
|        504 | Anjali       |

SELECT Projects.Project_Name, Managers.Manager_Name
FROM Projects
FULL JOIN Managers
ON Projects.Manager_ID = Managers.Manager_ID;

| Project_Name | Manager_Name |
| ------------ | ------------ |
| Website      | Meera        |
| Mobile App   | Vikram       |
| Database     | NULL         |
| NULL         | Anjali       |

```python
import sqlite3

# Create an in-memory database
db = sqlite3.connect(":memory:")
cursor = db.cursor()

# Create Projects table
cursor.execute("""
    CREATE TABLE Projects (
        project_id INTEGER PRIMARY KEY,
        project_name TEXT,
        manager_id INTEGER
    )
""")

# Create Managers table
cursor.execute("""
    CREATE TABLE Managers (
        manager_id INTEGER PRIMARY KEY,
        manager_name TEXT
    )
""")

# Add project data
project_data = [
    (201, "Website", 501),
    (202, "Mobile App", 502),
    (203, "Database", 503)
]

cursor.executemany("""
    INSERT INTO Projects
    VALUES (?, ?, ?)
""", project_data)

# Add manager data
manager_data = [
    (501, "Meera"),
    (502, "Vikram"),
    (504, "Anjali")
]

cursor.executemany("""
    INSERT INTO Managers
    VALUES (?, ?)
""", manager_data)

# Reproduce FULL JOIN behavior
cursor.execute("""
    SELECT Projects.project_name, Managers.manager_name
    FROM Projects
    LEFT JOIN Managers
        ON Projects.manager_id = Managers.manager_id

    UNION

    SELECT Projects.project_name, Managers.manager_name
    FROM Managers
    LEFT JOIN Projects
        ON Projects.manager_id = Managers.manager_id
""")

results = cursor.fetchall()

print("Projects and Managers:")

for project, manager in results:
    print(project, "->", manager)

db.close()
```

CROSS JOIN:

A CROSS JOIN combines every row from the first table with every row from the second table.

Unlike INNER JOIN, LEFT JOIN, or RIGHT JOIN, a CROSS JOIN does not need a matching column or ON condition.

EXAMPLE:

Cloting company wants to know every possible colour size combination:

| Color | Size   |
| ----- | ------ |
| Red   | Small  |
| Red   | Medium |
| Blue  | Small  |
| Blue  | Medium |
| Green | Small  |
| Green | Medium |

```python
import sqlite3

# Create an in-memory database
database = sqlite3.connect(":memory:")
cursor = database.cursor()

# Create Colors table
cursor.execute("""
    CREATE TABLE Colors (
        color_id INTEGER PRIMARY KEY,
        color_name TEXT
    )
""")

# Create Sizes table
cursor.execute("""
    CREATE TABLE Sizes (
        size_id INTEGER PRIMARY KEY,
        size_name TEXT
    )
""")

# Insert color data
color_data = [
    (1, "Red"),
    (2, "Blue"),
    (3, "Green")
]

cursor.executemany("""
    INSERT INTO Colors
    VALUES (?, ?)
""", color_data)

# Insert size data
size_data = [
    (1, "Small"),
    (2, "Medium")
]

cursor.executemany("""
    INSERT INTO Sizes
    VALUES (?, ?)
""", size_data)

# Use CROSS JOIN
cursor.execute("""
    SELECT Colors.color_name, Sizes.size_name
    FROM Colors
    CROSS JOIN Sizes
""")

combinations = cursor.fetchall()

print("Possible color and size combinations:")

for color, size in combinations:
    print(color, "-", size)

database.close()
```

The result contains:

3 colors × 2 sizes = 6 combinations

In Simple Words:

CROSS JOIN creates every possible combination between two tables. 
It does not require a matching condition and produces a Cartesian product of the two tables.

UNION:

Unlike a JOIN, which combines columns from related tables, UNION combines the rows returned by separate queries.

EXAMPLE:
Customers from 2 different branches:

CHENNAI CUSTOMERS TABLE:

| Customer_ID | Customer_Name |
| ----------: | ------------- |
|         301 | Asha          |
|         302 | Rohit         |
|         303 | Neha          |


PUNE CUSTOMERS:

| Customer_ID | Customer_Name |
| ----------: | ------------- |
|         401 | Karan         |
|         402 | Meena         |
|         303 | Neha          |

we want one list containing customers from both branches:

SELECT Customer_ID, Customer_Name
FROM Chennai_Customers

UNION

SELECT Customer_ID, Customer_Name
FROM Pune_Customers;

OUTPUT:

| Customer_ID | Customer_Name |
| ----------: | ------------- |
|         301 | Asha          |
|         302 | Rohit         |
|         303 | Neha          |
|         401 | Karan         |
|         402 | Meena         |

```python
import sqlite3

# Create an in-memory database
connection = sqlite3.connect(":memory:")
cursor = connection.cursor()

# Create two branch tables
cursor.execute("""
    CREATE TABLE North_Orders (
        order_id INTEGER,
        customer_name TEXT
    )
""")

cursor.execute("""
    CREATE TABLE South_Orders (
        order_id INTEGER,
        customer_name TEXT
    )
""")

# Add orders from the North branch
north_orders = [
    (501, "Asha"),
    (502, "Rohit"),
    (503, "Neha")
]

cursor.executemany("""
    INSERT INTO North_Orders
    VALUES (?, ?)
""", north_orders)

# Add orders from the South branch
south_orders = [
    (601, "Karan"),
    (602, "Meena"),
    (503, "Neha")
]

cursor.executemany("""
    INSERT INTO South_Orders
    VALUES (?, ?)
""", south_orders)

connection.commit()

# Combine the two query results
cursor.execute("""
    SELECT order_id, customer_name
    FROM North_Orders

    UNION

    SELECT order_id, customer_name
    FROM South_Orders
""")

combined_orders = cursor.fetchall()

print("Combined orders:")

for order in combined_orders:
    print(order)

connection.close()
```

Suppose we want to find employees who earn more than the average salary.

First, we need to calculate the average salary:

SELECT AVG(Salary)
FROM Employees;

Instead of manually putting that number into another query, we can use a subquery:

SELECT Employee_Name, Salary

```python
import sqlite3

# Create an in-memory database
connection = sqlite3.connect(":memory:")
cursor = connection.cursor()

# Create Employees table
cursor.execute("""
    CREATE TABLE Employees (
        employee_id INTEGER PRIMARY KEY,
        employee_name TEXT,
        salary INTEGER,
        team TEXT
    )
""")

# Insert employee information
employee_data = [
    (801, "Divya", 55000, "Design"),
    (802, "Sameer", 72000, "Engineering"),
    (803, "Kavya", 68000, "Marketing"),
    (804, "Nitin", 85000, "Engineering"),
    (805, "Farah", 62000, "Design")
]

cursor.executemany("""
    INSERT INTO Employees
    VALUES (?, ?, ?, ?)
""", employee_data)

connection.commit()

# Use a subquery to find employees
# earning more than the average salary
cursor.execute("""
    SELECT employee_name, salary
    FROM Employees
    WHERE salary > (
        SELECT AVG(salary)
        FROM Employees
    )
""")

results = cursor.fetchall()

print("Employees earning above average:")

for employee, salary in results:
    print(employee, "-", salary)

connection.close()
```

```python
import sqlite3

# Create an in-memory database
connection = sqlite3.connect(":memory:")
cursor = connection.cursor()

# Create Books table
cursor.execute("""
    CREATE TABLE Books (
        book_id INTEGER PRIMARY KEY,
        title TEXT,
        category TEXT
    )
""")

# Create Borrowing table
cursor.execute("""
    CREATE TABLE Borrowing (
        borrow_id INTEGER PRIMARY KEY,
        book_id INTEGER,
        days_borrowed INTEGER
    )
""")

# Insert book data
book_data = [
    (901, "Python Basics", "Programming"),
    (902, "SQL Guide", "Database"),
    (903, "Data Science", "Programming"),
    (904, "Networking", "IT")
]

cursor.executemany("""
    INSERT INTO Books
    VALUES (?, ?, ?)
""", book_data)

# Insert borrowing data
borrowing_data = [
    (1, 901, 12),
    (2, 902, 5),
    (3, 903, 18),
    (4, 904, 7),
    (5, 901, 8)
]

cursor.executemany("""
    INSERT INTO Borrowing
    VALUES (?, ?, ?)
""", borrowing_data)

connection.commit()

# Use a Common Table Expression
cursor.execute("""
    WITH BookAverages AS (
        SELECT
            book_id,
            AVG(days_borrowed) AS average_days
        FROM Borrowing
        GROUP BY book_id
    )
    SELECT
        Books.title,
        BookAverages.average_days
    FROM Books
    JOIN BookAverages
        ON Books.book_id = BookAverages.book_id
    WHERE BookAverages.average_days > 10
""")

results = cursor.fetchall()

print("Books with average borrowing time above 10 days:")

for title, average_days in results:
    print(title, "-", average_days)

connection.close()
```

A window function performs a calculation across a group of related rows while keeping each individual row in the result.

This is the main difference from GROUP BY.

With GROUP BY, several rows are usually combined into one summary row. With a window function, the original rows remain visible, and the calculated value is added alongside them.

A window function calculates something across related rows without collapsing those rows into a single row.

Window functions are commonly used for:

1. Ranking records

```python
import sqlite3

# Create an in-memory database
connection = sqlite3.connect(":memory:")
cursor = connection.cursor()

# Create MatchScores table
cursor.execute("""
    CREATE TABLE MatchScores (
        match_id INTEGER PRIMARY KEY,
        player TEXT,
        team TEXT,
        points INTEGER
    )
""")

# Insert tournament data
score_data = [
    (1, "Aarav", "Tigers", 80),
    (2, "Isha", "Tigers", 65),
    (3, "Kabir", "Tigers", 90),
    (4, "Riya", "Falcons", 75),
    (5, "Dev", "Falcons", 88),
    (6, "Tara", "Falcons", 70)
]

cursor.executemany("""
    INSERT INTO MatchScores
    VALUES (?, ?, ?, ?)
""", score_data)

connection.commit()

# Use a window function to rank players
cursor.execute("""
    SELECT
        player,
        team,
        points,
        RANK() OVER (
            PARTITION BY team
            ORDER BY points DESC
        ) AS team_rank
    FROM MatchScores
    ORDER BY team, team_rank
""")

results = cursor.fetchall()

print("Player rankings:")

for player, team, points, rank in results:
    print(player, "-", team, "-", points, "-", rank)

connection.close()
```
